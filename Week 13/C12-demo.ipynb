## 1. Determine PC

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# ----------------------------
# TẠO DỮ LIỆU 2D CÓ TƯƠNG QUAN
# ----------------------------
np.random.seed(0)  # Khởi tạo hạt giống ngẫu nhiên để kết quả có thể tái lập
mean = [0, 0]      # Trung bình của 2 chiều dữ liệu
# Ma trận hiệp phương sai, thể hiện tương quan giữa 2 chiều
'''
cov = [[var_x, cov_xy],
       [cov_xy, var_y]]
'''
cov = [[4, 3.5],   
       [3.5, 4]]
data = np.random.multivariate_normal(mean, cov, 300)  # Sinh 300 điểm dữ liệu 2D theo phân phối chuẩn đa biến

# ----------------------------
# PCA
# ----------------------------
pca = PCA(n_components=2)  # Khởi tạo PCA giữ 2 thành phần chính
pca.fit(data)              # Học PCA từ dữ liệu
pc1 = pca.components_[0]   # Thành phần chính 1 (PC1)
pc2 = pca.components_[1]   # Thành phần chính 2 (PC2)

# ----------------------------
# TÍNH ĐƯỜNG THẲNG THEO PC1/PC2
# Đường kéo dài từ biên min đến biên max theo hướng thành phần chính
# ----------------------------
# Chiếu dữ liệu lên PC1 và PC2
proj1 = data @ pc1  # Tọa độ dữ liệu trên PC1
proj2 = data @ pc2  # Tọa độ dữ liệu trên PC2

# Xác định điểm đầu và điểm cuối của đường PC1 dựa trên biên min/max
pc1_start = np.mean(data, axis=0) + pc1 * proj1.min()  # Điểm biên min theo PC1
pc1_end   = np.mean(data, axis=0) + pc1 * proj1.max()  # Điểm biên max theo PC1

# Xác định điểm đầu và điểm cuối của đường PC2 dựa trên biên min/max
pc2_start = np.mean(data, axis=0) + pc2 * proj2.min()  # Điểm biên min theo PC2
pc2_end   = np.mean(data, axis=0) + pc2 * proj2.max()  # Điểm biên max theo PC2

line_width = 2  # Độ dày đường thẳng

# ----------------------------
# HÌNH 1: ĐÁM MÂY 2D + PC1, PC2
# ----------------------------
plt.figure(figsize=(8, 8))  # Tạo hình vuông để PC1 và PC2 hiển thị vuông góc trực quan
plt.scatter(data[:, 0], data[:, 1], c="#d48c00", marker='x', label="Dữ liệu gốc")  # Vẽ các điểm dữ liệu màu cam

# Vẽ đường thẳng PC1 & PC2 từ biên min đến biên max
plt.plot([pc1_start[0], pc1_end[0]], [pc1_start[1], pc1_end[1]], color='red', linewidth=line_width, label="PC1")
plt.plot([pc2_start[0], pc2_end[0]], [pc2_start[1], pc2_end[1]], color='blue', linewidth=line_width, label="PC2")

# Thêm nhãn PC1/PC2
plt.text(pc1_end[0]*1.05, pc1_end[1]*1.05, "PC1", color='red', fontsize=12)
plt.text(pc2_end[0]*1.05, pc2_end[1]*1.05, "PC2", color='blue', fontsize=12)

plt.title("Đám mây dữ liệu 2D và thành phần chính (PC1 & PC2) từ biên đến biên")
plt.xlabel("X1")
plt.ylabel("X2")
plt.grid(True)          # Hiển thị lưới
plt.axis('equal')       # Giữ tỉ lệ trục X=Y, giúp PC1 và PC2 vuông góc trực quan
plt.legend()
plt.show()

# ----------------------------
# HÌNH 2: Chiếu dữ liệu lên PC1
# ----------------------------
plt.figure(figsize=(8, 8))
plt.scatter(data[:, 0], data[:, 1], c="#d48c00", marker='x', label="Dữ liệu gốc")  # Vẽ dữ liệu gốc
plt.scatter(np.outer(proj1, pc1)[:,0], np.outer(proj1, pc1)[:,1], c='skyblue', marker='s', label="Chiếu lên PC1")  # Vẽ dữ liệu chiếu lên PC1

# Vẽ đường PC1 từ biên min đến biên max
plt.plot([pc1_start[0], pc1_end[0]], [pc1_start[1], pc1_end[1]], c="red", linewidth=line_width, label="PC1")

plt.title("Chiếu các điểm 2D lên PC1 (2D -> 1D) và khôi phục lại")
plt.xlabel("X1")
plt.ylabel("X2")
plt.grid(True)
plt.axis('equal')  # Giữ tỉ lệ trục
plt.legend()
plt.show()
```

## 2. Reducing the dimensionality of the Iris set to 2D

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA

# ----------------------------
# TẢI DỮ LIỆU IRIS
# ----------------------------
iris = load_iris()
X = iris.data       # Ma trận dữ liệu: 150 mẫu x 4 đặc trưng
y = iris.target     # Nhãn lớp (0,1,2)
target_names = iris.target_names  # Tên các lớp: ['setosa', 'versicolor', 'virginica']

# ----------------------------
# GIẢM SỐ CHIỀU BẰNG PCA
# ----------------------------
pca = PCA(n_components=2)  # Giảm xuống 2 chiều
X_r = pca.fit_transform(X)  # Chiếu dữ liệu xuống 2 chiều

# ----------------------------
# VẼ SCATTER PLOT
# ----------------------------
plt.figure(figsize=(8, 6))

colors = ['red', 'green', 'blue']  # Màu cho 3 lớp
for color, i, target_name in zip(colors, [0, 1, 2], target_names):
    plt.scatter(X_r[y == i, 0], X_r[y == i, 1], color=color, label=target_name)

plt.title("Giảm số chiều tập Iris xuống 2 chiều bằng PCA")
plt.xlabel("PC1")  # Thành phần chính 1
plt.ylabel("PC2")  # Thành phần chính 2
plt.legend()
plt.grid(True)
plt.show()
```

## 3. Reducing the dimensionality of the Iris set to 3D

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.decomposition import PCA
from mpl_toolkits.mplot3d import Axes3D  # Thư viện hỗ trợ vẽ 3D

# ----------------------------
# TẢI DỮ LIỆU IRIS
# ----------------------------
iris = load_iris()
X = iris.data       # Ma trận dữ liệu: 150 mẫu x 4 đặc trưng
y = iris.target     # Nhãn lớp (0,1,2)
target_names = iris.target_names  # Tên các lớp: ['setosa', 'versicolor', 'virginica']

# ----------------------------
# GIẢM SỐ CHIỀU BẰNG PCA
# ----------------------------
pca = PCA(n_components=3)  # Giảm xuống 3 chiều
X_r = pca.fit_transform(X)  # Chiếu dữ liệu xuống 3 chiều

# ----------------------------
# VẼ SCATTER PLOT 3D
# ----------------------------
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')  # Khởi tạo biểu đồ 3D

colors = ['red', 'green', 'blue']  # Màu cho 3 lớp
for color, i, target_name in zip(colors, [0, 1, 2], target_names):
    ax.scatter(X_r[y == i, 0], X_r[y == i, 1], X_r[y == i, 2], 
               color=color, label=target_name, s=50)  # s=50 là kích thước điểm

# Thiết lập nhãn trục
ax.set_title("Giảm số chiều tập Iris xuống 3 chiều bằng PCA")
ax.set_xlabel("PC1")  # Thành phần chính 1
ax.set_ylabel("PC2")  # Thành phần chính 2
ax.set_zlabel("PC3")  # Thành phần chính 3

ax.legend()
plt.show()
```

## 4. Determine k

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.datasets import load_iris

# ----------------------------
# Bước 1: Load dữ liệu Iris
# ----------------------------
data = load_iris()
X = data.data  # Dữ liệu gồm 4 đặc trưng (features): sepal length, sepal width, petal length, petal width

# ----------------------------
# Bước 2: Áp dụng PCA
# ----------------------------
pca = PCA()    # Khởi tạo PCA (giữ tất cả các thành phần chính)
pca.fit(X)     # Học PCA trên dữ liệu X

# ----------------------------
# Bước 3: Lấy explained variance ratio và cumulative variance
# ----------------------------
explained_variance_ratio = pca.explained_variance_ratio_  # Tỷ lệ phương sai được giải thích bởi từng thành phần chính
cumulative_variance = np.cumsum(explained_variance_ratio)  # Phương sai tích lũy (cộng dồn)

# ----------------------------
# Bước 4: Vẽ biểu đồ explained variance và cumulative variance
# ----------------------------
plt.figure(figsize=(8,5))

# Vẽ cumulative variance (phương sai tích lũy) với marker 'o' và đường nét '--' màu xanh dương
plt.plot(range(1, len(explained_variance_ratio)+1), cumulative_variance, 
         marker='o', linestyle='--', color='b')

# Vẽ explained variance của từng PC với marker 'x' và đường nét '-' màu đỏ
plt.plot(range(1, len(explained_variance_ratio)+1), explained_variance_ratio, 
         marker='x', linestyle='-', color='r')

plt.title('Explained Variance and Cumulative Explained Variance')
plt.xlabel('Number of Principal Components')  # Trục X: số thành phần chính
plt.ylabel('Variance Ratio')                  # Trục Y: tỷ lệ phương sai
plt.xticks(range(1, len(explained_variance_ratio)+1))  # Đánh số các PC trên trục X
plt.grid(True)
plt.legend(['Cumulative Variance', 'Explained Variance per PC'])  # Chú thích
plt.show()

# ----------------------------
# Bước 5: Lựa chọn số PC tối ưu tại điểm "elbow"
# ----------------------------
# Thường chọn số PC sao cho cumulative_variance >= 0.9 (giải thích >=90% phương sai)
num_pc_optimal = np.argmax(cumulative_variance >= 0.9) + 1
print(f"Number of PCs k to retain (>=90% variance): {num_pc_optimal}")
```
