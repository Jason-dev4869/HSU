# Lab 02: LINEAR REGRESSION

```python
#libraries
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error
import pandas as pd

df = pd.read_csv('Housing.csv')
df.head()
```

### 1. Use only area to predict house price

```python
# Use only area to predict house price
df1 = df[['area', 'price']]

# normalize the data using MinMaxScaler
scaler = MinMaxScaler()
df1_scaled = scaler.fit_transform(df1)

# split the data into training and testing sets
X_df1 = df1_scaled[:, 0].reshape(-1, 1)  # area
y_df1 = df1_scaled[:, 1]  # price
X_train_df1, X_test_df1, y_train_df1, y_test_df1 = train_test_split(
    X_df1, y_df1, test_size=0.2, random_state=42)

# Write Linear Equation
model = LinearRegression()
model.fit(X_train_df1, y_train_df1)

#Predict the price of a house based on its area
predicted_prices_scaled = model.predict(X_test_df1)
# dataframe to store actual and predicted prices
results_df = pd.DataFrame({
    'Actual Price (scaled)': y_test_df1,
    'Predicted Price (scaled)': predicted_prices_scaled
})

#Compute MSE
y_pred_df1 = model.predict(X_test_df1)
mse_df1 = mean_squared_error(y_test_df1, y_pred_df1)
print(f'Mean Squared Error: {mse_df1}')

#Compute coefficient of determination R-squared score
r_squared_df1 = model.score(X_test_df1, y_test_df1)
print(f'R-squared score: {r_squared_df1}')

results_df
```

### 6. Use only area, bedrooms, bathrooms, stories, parking, airconditioning (yes->, no->0)
