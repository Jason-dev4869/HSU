```python
# libraries
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv('house_prices.csv')
#select only the columns we need
df = df[['LotArea', 'OverallQual', 'OverallCond',
         'YearBuilt', 'YearRemodAdd', 'TotalBsmtSF',
         '1stFlrSF', '2ndFlrSF', 'FullBath','HalfBath',
         'BedroomAbvGr', 'TotRmsAbvGrd', 'GarageArea', 'SalePrice']]
df.head()
```

### a) Compute the covariance matix and plot a heatmap of the coveriance matrix selected features using Seaborn package

```python
plt.figure(figsize=(12, 10))
sns.heatmap(df.cov(), annot=True, cmap='coolwarm')
plt.title('Covariance Matrix of House Prices Dataset')
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

df.cov()
```

### b) Plot the correlation heatmap of the of the correlation matrix of selected features using Seaborn package

```python
plt.figure(figsize=(12, 10))
sns.heatmap(df.corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of House Prices Dataset')
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()
```

### d) Plot scatterplots of 5 highly-corelated pairs of features and 5 non corelated pairs of features.

```python
five_highest_corr = df.corr()['SalePrice'].abs().sort_values(ascending=False).head(6)
most_correlated_features = five_highest_corr.index[1:]  # Exclude 'SalePrice' itself
for feature in most_correlated_features:
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=df[feature], y=df['SalePrice'])
    plt.title(f'Scatter Plot of {feature} vs SalePrice')
    plt.xlabel(feature)
    plt.ylabel('SalePrice')
    plt.tight_layout()
    plt.show()

print("Five Highest Correlated:")
five_highest_corr
```

```python
five_non_correlated = df.corr()['SalePrice'].abs().sort_values(ascending=True).head(5)

least_correlated_features = five_non_correlated.index
for feature in least_correlated_features:
    plt.figure(figsize=(8, 6))
    sns.scatterplot(x=df[feature], y=df['SalePrice'])
    plt.title(f'Scatter Plot of {feature} vs SalePrice')
    plt.xlabel(feature)
    plt.ylabel('SalePrice')
    plt.tight_layout()
    plt.show()

print("Five Non-Correlated:")
five_non_correlated
```

### e) Use pandas.plotting.scatter_matrix for the selected features. 

**Note**: pandas.plotting.scatter_matrixis a function within the pandas library that generates a matrix of scatter plots and histograms (or kernel density estimates) for a given DataFrame. This visualization tool is particularly useful for exploring relationships and distributions among multiple numerical variables in a dataset.

- **Scatter Plots**: For every unique pair of numerical columns in the input DataFrame, scatter_matrixcreates a scatter plot, allowing for the visual inspection of bivariate relationships and potential correlations. 

- **Histograms/KDEs**: Along the diagonal of the matrix, the function displays either histograms or Kernel Density Estimation (KDE) plots for each individual numerical variable. These plots illustrate the distribution of each variable. 

```python
#Selected Features using pandas.plotting.scatter_matrix
from pandas.plotting import scatter_matrix

selected_features = ['OverallQual', 'YearBuilt', 'TotalBsmtSF', '1stFlrSF', 'GarageArea']
scatter_matrix(df[selected_features], figsize=(12, 10), diagonal='kde')
plt.suptitle('Scatter Matrix of Selected Features')
plt.show()
```
