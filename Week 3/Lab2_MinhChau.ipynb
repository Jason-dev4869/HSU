```python
#Import library
import numpy as np
import pandas as pd

#Read database

df = pd.read_csv('Customers.csv')
df
```

1. Insert a new column called Initials that combine the first letter of FirstName and LastName (use .map() a Lambda function). For example: Joseph Martins -> JM

```python
df['Initials'] = df[['FirstName','LastName']].apply(lambda x: x['FirstName'][0] + x['LastName'][0], axis=1)
df['Initials'].head()
```

2. Find the number of duplicates and remove the duplicated, and only keep the first entry

```python
#Num of Duplicate
num_duplicates = df.duplicated().sum()
print('Number of duplicated rows: ', num_duplicates)

df[df.duplicated()]
```

```python
#Drop Duplicated
df.drop_duplicates(keep='first')
```

3. Find the number of profession

```python
print('Number of Profession: ', df['Profession'].nunique())
df['Profession'].value_counts()
```

4. Find the average income

```python
print("Average of Annual Income: ", df['Annual Income ($)'].mean(), "$")
```

5. Find the median income

```python
print("Median of Annual Income: ", df['Annual Income ($)'].median(), "$")
```

6. Extract info [FirstName, LastName, Gender, Age, Annual Income, Spending Score, Profession] of Male Engineers above 20 years old with income over 60000 $

```python
df.loc[(df.Age>20) & 
       (df.Gender == 'Male') & 
       (df.Profession == 'Engineer') & 
       (df['Annual Income ($)'] > 60000), 
       ['FirstName', 'LastName', 'Gender', 'Age', 'Annual Income ($)', 'Spending Score (1-100)', 'Profession']]
```

7. Extract info [FirstName, LastName, Gender, Age, Annual Income, Spending Score, Profession] of Female Artist between 20-40 years old with income below 40000 $

```python
df.loc[(df.Age.between(20,40,inclusive='both')) & 
       (df.Gender == 'Female') & 
       (df.Profession == 'Artist') & 
       (df['Annual Income ($)'] < 40000), 
       ['FirstName', 'LastName', 'Gender', 'Age', 'Annual Income ($)', 'Spending Score (1-100)', 'Profession']]
```

8. Extract those whose first name is Jac (use Series.str.contains())

```python
df.loc[df.FirstName.str.contains('Jac', case=False, na=False)]
```

9.
    Extract those whose spending score is above 70
    
    Extract males whose spending score is above 70
    
    Extract females whose spending score is above 70

```python
#All Gender Spending Score Above 70
df.loc[df['Spending Score (1-100)'] > 70]
```

```python
#All Male Spending Score Above 70
male_ss70 = df.loc[(df['Spending Score (1-100)'] > 70) &
   (df.Gender == 'Male')]
male_ss70
```

```python
#All Female Spending Score Above 70
female_ss70 = df.loc[(df['Spending Score (1-100)'] > 70) &
   (df.Gender == 'Female')]
female_ss70
```

10. Find the percentage of Male with spending score above 70 vs percentage of Female with spending score above 70

```python
#Total Number of each Gender Entire DataFrame
total_male = df.loc[df.Gender == 'Male'].shape[0]
total_female = df.loc[df.Gender == 'Female'].shape[0]

#Percentage of each Gender with Spending Score Above 70
male_percent = (male_ss70.shape[0] / total_male) * 100
female_percent = (female_ss70.shape[0] / total_female) * 100

print(f"Percentage of Male with Spending Score > 70: {male_percent:.2f}%")
print(f"Percentage of Female with Spending Score > 70: {female_percent:.2f}%")
```
