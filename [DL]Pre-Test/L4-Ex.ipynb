## Bài 5. Tính gradient bằng `tf.GradientTape()`

Yêu cầu:

1. Tạo biến `w = 3.0` bằng `tf.Variable`.
2. Xét hàm `y = w^2 + 2w + 1`.
3. Dùng `tf.GradientTape()` để tính gradient `dy/dw`.
4. In ra giá trị của `y` và gradient thu được.

```python
# TODO: Tạo biến w
w = tf.Variable(3.0)

# TODO: Dùng GradientTape để tính y và gradient dy/dw
with tf.GradientTape() as tape:
    y = w**2 + 2*w + 1

# TODO: In ra giá trị y
print("Value of y:", y.numpy())

# TODO: In ra gradient dy/dw
print("Gradient dy/dw:", tape.gradient(y, w).numpy())
```

## Bài 6. Mô phỏng gradient descent đơn giản

Yêu cầu:

Xét hàm mất mát: $loss = (w - 4.0)^2$

Thực hiện gradient descent trong 10 bước để cập nhật `w` sao cho `w` tiến dần về 4.

Yêu cầu cụ thể:

1. Khởi tạo `w = 0.0`
2. Đặt `learning_rate = 0.1`
3. Lặp 10 bước
4. Mỗi bước in ra: số bước, giá trị `w`, giá trị `loss`, gradient

```python
# TODO: Khởi tạo w và learning_rate
w = tf.Variable(0.0)
learning_rate = 0.1

# TODO: Viết vòng lặp gradient descent 10 bước
for step in range(10):
    with tf.GradientTape() as tape:
        loss = (w - 4.0)**2
    dy_dw = tape.gradient(loss, w)
    w.assign_sub(learning_rate * dy_dw)
    print(f"Step {step+1}: w = {w.numpy()}, loss = {loss.numpy()}")
```

## Bài 7. Xây dựng mô hình Keras đơn giản

Yêu cầu:

Tạo mô hình `Sequential` gồm:

1. Một layer `Dense` có 16 neuron, activation=`relu`, input shape = `(8,)`
2. Một layer output `Dense` có 1 neuron, activation=`sigmoid`
3. Hiển thị `model.summary()`

```python
# TODO: Tạo mô hình Sequential theo yêu cầu
model = keras.Sequential([
    layers.Dense(16, activation='relu', input_shape=(8,)),
    layers.Dense(1, activation='sigmoid')
])

# TODO: Hiển thị summary của mô hình
model.summary()
```

## Bài 8. Tạo dữ liệu giả cho bài toán phân loại nhị phân

Yêu cầu:

1. Dùng NumPy tạo `x_train` gồm 200 mẫu, mỗi mẫu có 8 đặc trưng ngẫu nhiên.
2. Tạo `y_train` theo quy tắc: nếu tổng 8 đặc trưng > 4 thì gán nhãn 1, ngược lại nhãn 0.
3. Tạo tương tự `x_test` gồm 50 mẫu và `y_test`.
4. In shape của `x_train`, `y_train`, `x_test`, `y_test`.

```python
# TODO: Đặt random seed để kết quả có thể tái lập
np.random.seed(42)

# TODO: Tạo x_train và y_train
x_train = np.random.rand(200, 8)
y_train = (np.sum(x_train, axis=1) > 4).astype('float32')

# TODO: Tạo x_test và y_test
x_test = np.random.rand(50, 8)
y_test = (np.sum(x_test, axis=1) > 4).astype('float32')

# TODO: In shape của các tập dữ liệu
print("Shape of x_train:", x_train.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of x_test:", x_test.shape)
print("Shape of y_test:", y_test.shape)
```

## Bài 9. Compile và huấn luyện mô hình

Yêu cầu:

1. Compile model với:
   - optimizer = `rmsprop`
   - loss = `binary_crossentropy`
   - metrics = `accuracy`
2. Train mô hình với:
   - epochs = 5
   - batch_size = 16
   - validation_split = 0.2
3. Lưu lịch sử train vào biến `history`.

```python
# TODO: Compile mô hình
model.compile(
    optimizer='rmsprop',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# TODO: Huấn luyện mô hình và lưu vào biến history
history = model.fit(
    x_train, y_train,
    epochs=5,
    batch_size=16,
    validation_split=0.2
)
```

## Bài 10. Đánh giá mô hình và dự đoán

Yêu cầu:

1. Đánh giá mô hình trên tập test.
2. In ra `test_loss` và `test_accuracy`.
3. Dự đoán xác suất cho 5 mẫu đầu tiên của `x_test`.
4. In kết quả dự đoán.
5. Hãy cho biết: nếu output sigmoid >= 0.5 thì thuộc lớp nào?

```python
# TODO: Đánh giá mô hình trên tập test
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)

# TODO: In test_loss và test_accuracy
print('Test loss =', round(float(test_loss), 4))
print('Test accuracy =', round(float(test_acc), 4))

# TODO: Dự đoán xác suất cho 5 mẫu đầu tiên
pred = model.predict(x_test[:5])

# TODO: In kết quả dự đoán
print('Predicted probabilities for first 5 test samples:')
print(pred)
```

    `tf.Variable` là một đối tượng có thể thay đổi giá trị trong quá trình huấn luyện, trong khi tensor hằng (constant) có giá trị cố định và không thể thay đổi sau khi được tạo ra.

    Điều này rất hữu ích trong quá trình huấn luyện mô hình, vì nó giúp cập nhật trọng số của mô hình dựa trên gradient descent hoặc các thuật toán tối ưu hóa khác.

- Câu 4:

    `fit()` là bước thực sự huấn luyện mô hình trên dữ liệu. Nó nhận vào dữ liệu huấn luyện, số epoch, batch size và các tham số khác để bắt đầu quá trình học. `fit()` sẽ sử dụng cấu hình đã thiết lập trong `compile()` để cập nhật trọng số của mô hình dựa trên dữ liệu huấn luyện.

## Bài 12. Bài tập mở rộng

Yêu cầu:

Hãy thay đổi một trong các yếu tố sau và quan sát kết quả huấn luyện:

- Số neuron ở hidden layer: 16 → 32
- Số epoch: 5 → 10
- Batch size: 16 → 8 hoặc 32

Viết nhận xét ngắn: Việc thay đổi đó có ảnh hưởng gì đến kết quả train hoặc test?

```python
# TODO: Thử thay đổi một tham số huấn luyện hoặc kiến trúc mô hình
# Gợi ý: tạo lại model, compile lại, train lại và so sánh kết quả

np.random.seed(42)
x_train = np.random.rand(200, 8)
y_train = (np.sum(x_train, axis=1) > 4).astype('float32')
x_test = np.random.rand(50, 8)
y_test = (np.sum(x_test, axis=1) > 4).astype('float32')

model = keras.Sequential([
    layers.Dense(32, activation='relu', input_shape=(8,)),
    layers.Dense(1, activation='sigmoid')
])
model.summary()

model.compile(
    optimizer='rmsprop',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

history = model.fit(
    x_train, y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)

test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)
print('Test loss =', round(float(test_loss), 4))
print('Test accuracy =', round(float(test_acc), 4))

pred = model.predict(x_test[:5])
print('Predicted probabilities for first 5 test samples:')
print(pred)
```

### Nhận xét của sinh viên

    Việc tăng số neuron ở hidden layer từ 16 lên 32 giúp mô hình có khả năng học được nhiều đặc trưng hơn từ dữ liệu, điều này có thể cải thiện độ chính xác của mô hình. 

    Tăng số epoch từ 5 lên 10 cho phép mô hình có thêm thời gian để học và tối ưu hóa, điều này cũng có thể dẫn đến kết quả tốt hơn. 

    Thay đổi batch size từ 16 lên 32 có thể giúp quá trình huấn luyện ổn định hơn và nhanh hơn, nhưng cũng có thể làm giảm khả năng tổng quát hóa của mô hình nếu batch size quá lớn. 

    Tổng thể, những thay đổi này thường sẽ cải thiện hiệu suất của mô hình trên tập test, nhưng cũng cần phải thử nghiệm để tìm ra cấu hình tốt nhất.
