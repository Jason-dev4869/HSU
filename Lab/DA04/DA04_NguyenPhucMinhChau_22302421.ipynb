Exercise 1: Creating Arrays with `arange`

Question: Use `numpy.arange` to create a 1D array of even numbers from 10 to 20 (inclusive of 10, exclusive of 20) with a step of 2. Print the array and its data type.

```python
import numpy as np

A = np.arange(10, 20, 2)
print("10 to 20 (step 2) Array:",A)
print("Array type:",A.dtype)
```

Exercise 3: Arithmetic Operations

Question: Create two 1D arrays `a = [1, 2, 3]` and `b = [4, 5, 6]`. Perform element-wise addition, multiplication, and exponentiation. Also compute the matrix dot product.

```python
a = np.array([1,2,3])
b = np.array([4,5,6])

print("Addition of a anb b:", np.add(a,b))
print("Multiplication of a and b:", np.multiply(a,b))
print("Exponentiation of a and b:", np.power(a,b))
print("Matrix dot:", np.dot(a,b))
```

Exercise 4: Indexing and Slicing

Question: Given the 2D array `matrix = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])`, extract:

1. The element in the second row, third column.

2. The first two rows.

3. All elements from the last column.

```python
matrix = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

print("The element in the second row, third column:", matrix[1,2])
print("The first two rows:",matrix[:2])
print("All elements from the last column:", matrix[:,-1])
```

Exercise 6: Aggregate Functions

Question: Given the array `arr = np.array([15, 22, 8, 19, 27])`, compute the sum, mean, maximum, and standard deviation.

```python
arr = np.array([15, 22, 8, 19, 27])

print("Sum of array:",np.sum(arr))
print("Mean of array:",np.mean(arr))
print("Maximum of array:",np.max(arr))
print("Standard deviation of array:",np.std(arr))
```

Exercise 7: Boolean Indexing

Question: From the array `arr = np.array([5, 12, 8, 3, 15, 7])`, extract all elements greater than 8.

```python
arr = np.array([5, 12, 8, 3, 15, 7])

print("Elements greater than 8:", arr[arr > 8])
```

Exercise 8: Shape Manipulation

Question: Create a 1D array with values from 0 to 11. Reshape it into a 3×4 matrix. Then flatten it back using `ravel()`.

```python
arr = np.arange(0,12)
print(arr)

arr1 = np.reshape(arr,(3,4))
print("\nReshape array to 3x4:\n", arr1)
print("\nFlatten array back:", np.ravel(arr1))
```

Exercise 9: Stacking Arrays

Question: Create two 2×2 arrays `a = [[1, 2], [3, 4]]` and `b = [[5, 6], [7, 8]]`. Stack them vertically and horizontally.

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])

print("Vertical stack 2 array:\n", np.vstack([a,b]))
print("Horizontal stack 2 array:\n", np.hstack([a,b]))
```

Exercise 10: Structured Arrays

Question: Create a structured array to store information about students: `name` (string of max 10 chars), `age` (int), and `grade` (float). Add at least two entries and print the names and ages.

```python
dtype = np.dtype([
    ('name',  'U10'),   # Unicode string, max 10 chars
    ('age',   'i4'),    # 32-bit integer
    ('grade', 'f4')     # 32-bit float
])

students = np.array([
    ('Alice',   20, 88.5),
    ('Bob',     22, 91.0)
], dtype=dtype)

print(students['name'])
print(students['age'])
```
