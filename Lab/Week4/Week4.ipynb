### Lab 1: Setting Up the Environment

```python
from ultralytics import YOLO
import cv2
```

### Lab 2: Load a YOLO Model and Run Inference

```python
# Load the pre-trained YOLOv8 nano model
model = YOLO("yolov8n.pt") #download automatically on first run

# Read a still image
img = cv2.imread("scene.jpg")

# Run detection
result = model(img)
```

### Lab 3: Reading Boxes and Classes

```python
for r in result:
    for box in r.boxes:         
        cls_id = int(box.cls[0])        # class index (e.g., 0 = person)         
        conf   = float(box.conf[0])     # confidence score (e.g., 0.87)         
        x1, y1, x2, y2 = box.xyxy[0]   # box corners in pixels         
        
        label = model.names[cls_id]     # human-readable class name         
        print(f"{label}: {conf:.2f}  box=[{x1:.0f},{y1:.0f},{x2:.0f},{y2:.0f}]")
```

### Lab 4: Computing the Box Center

```python
# Compute center of the bounding box
u = int((x1 + x2) / 2)
v = int((y1 + y2) / 2)

#Draw a red dot at the center on the image
cv2.circle(img, (u, v), 5, (0, 0, 255), -1)

#Optionally draw the bounding box itself
cv2.rectangle(img, (int(x1), int(y1)), (int(x2), int(y2)), (0, 255, 0), 2)
```

### Lab 5: Real-Time Webcam Detection

```python
cap = cv2.VideoCapture(0) # 0 = default webcam

while True:
    ret, frame = cap.read()
    if not ret:
        break

    results = model(frame)

    for r in results:
        for box in r.boxes:
            cls_id = int(box.cls[0])
            conf = float(box.conf[0])
            x1, x2, y1, y2 = box.xyxy[0]
            u = int((x1 + x2) / 2)
            v = int((y1 + y2) / 2)
            # Draw box, label, confidence and center point
            cv2.circle(frame, (u, v), 6, (0, 0, 255), -1)

    cv2.imshow("YOLO Detection", frame)
    if cv2.waitKey(1) != -1: break
cap.release
cv2.destroyAllWindows()
```
