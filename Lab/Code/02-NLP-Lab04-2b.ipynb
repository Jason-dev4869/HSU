### Tách câu (Sentence Segmentation) - Breaking text into individual sentences

```python
from underthesea import sent_tokenize

text = """Giữa cái nắng hạ êm đềm trong giấc mộng, anh đã được gặp lại em sau cả trăm năm. 
Nét son kiều diễm như ngày trước, quốc sắc thiên hương một ánh nhìn. Xuân hạ thu đông vẫn chuyển đổi."""
sentences = sent_tokenize(text)

for i,s in enumerate(sentences, 1): print(f"Câu {i}:{s}")
```

### Chuẩn hóa văn bản (Text Normalization) - Standardizing textual data representation and address conversion

```python
from underthesea import text_normalize

text_normalize('Địa chỉ: 15 Nguyễn Huệ, Q.1, tp.hcm')
```

### Gắn thẻ (Tagging) - Word segmentation, POS tagging, chunking, dependency parsing

```python
from underthesea import word_tokenize, pos_tag, chunk, dependency_parse

s = "Tôi yêu Việt Nam và đang học trí tuệ nhân tạo."

word_tokenize(s)          # tách từ
```

```python
pos_tag(s)               # từ loại
```

```python
chunk(s)                 # cụm từ
```

```python
dependency_parse(s) #bị lỗi GPU
```

### Nhận dạng thực thể định danh (Named Entity Recognition) - Identifying named entities (e.g., names, locations)

```python
from underthesea import ner

text = "Chủ tịch Nguyễn Phú Trọng gặp Tổng thống Joe Biden tại Washington."
result = ner(text)

# mỗi phần tử: (token, pos, chunk, ner_label)
for token, pos, chunk_tag, ner_tag in result:
    if ner_tag != 'O':
        print(token, '->', ner_tag)
```

### Phân lớp (Classification) - Text classification and sentiment analysis

```python
from underthesea import classify

classify("Bộ Y tế cảnh báo về dịch bệnh mới bùng phát.")
```

```python
from underthesea import sentiment

sentiment("Sản phẩm rất tốt, tôi rất hài lòng!")
```

### Xác định ngôn ngữ (Lang Detect) - Identifying the Language of Text

```python
from underthesea import lang_detect

print("Ngôn ngữ: ", lang_detect("Chào mừng bạn đến với Việt Nam."))
print("Ngôn ngữ: ", lang_detect("Hello, welcome to Vietnam."))
print("Ngôn ngữ: ", lang_detect("Bonjour, bienvenue au Vietnam."))
print("Ngôn ngữ: ", lang_detect("日本語のテキストです。"))
```

### Dịch tự động (Translation) - Translating Vietnamese text to English

```python
from underthesea import translate

sentences = [
    "Trí tuệ nhân tạo đang thay đổi thế giới.",
    "Tôi muốn học về điện toán lượng tử.",
    "Xin chào, rất vui được gặp bạn.",
]

for vi in sentences:
    en = translate(vi)          # mặc định vi → en, không cần truyền source/target
    print(f"VI: {vi}\nEN: {en}\n")
```
