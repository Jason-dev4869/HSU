## Bài tập 3: Tìm hiểu và demo minh họa 2 trình gán nhãn từ loại (POS tagger) dành cho tiếng Anh.

Có hai mô hình NLP phổ biến dành cho tiếng Anh nhất vẫn được cho ứng dụng vào tới thời điểm hiện có trình gán nhãn (POS tagger) là NLTK (Nature Language Toolkit) và spaCy.

1. NLTK (Natural Language Toolkit):
    
    NLTK là một thư viện Python mạnh mẽ và phổ biến cho xử lý ngôn ngữ tự nhiên bằng tiếng Anh. Nó cung cấp nhiều công cụ và tài nguyên cho các tác vụ NLP, bao gồm pos tagging.NLTK sử dụng một số mô hình khác nhau để thực hiện pos tagging, bao gồm cả mô hình dựa trên quy tắc và mô hình học máy.
    
    Để sử dụng pos tagging trong NLTK, sử dụng hàm `pos_tag` sau khi đã token hóa câu. Kết quả sẽ là một danh sách các tuple, mỗi tuple chứa một từ và nhãn pos tương ứng.

2. SpaCy:

    SpaCy là một thư viện NLP hiện đại và hiệu quả, được thiết kế để xử lý văn bản lớn và phức tạp. SpaCy sử dụng một mô hình học sâu để thực hiện pos tagging, và nó thường cho kết quả chính xác hơn so với các mô hình dựa trên quy tắc.
    
    Để sử dụng pos tagging trong SpaCy, cần tải mô hình ngôn ngữ (ví dụ: "en_core_web_sm") và sau đó xử lý văn bản bằng cách tạo một đối tượng `Doc`. Mỗi token trong đối tượng `Doc` sẽ có thuộc tính `pos_` chứa nhãn pos tương ứng.

### Demo POS tagger NLTK

```python
from nltk import word_tokenize, pos_tag

sentence = "The quick brown fox jumps over the lazy dog."
tokens = word_tokenize(sentence)
pos_tags = pos_tag(tokens)
pos_tags
```

### Demo POS tagger SpaCy

```python
# Demo pos tagging with spaCy
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp(sentence)

for token in doc:
    print(f"{token.text}: {token.pos_}")
```

## Bài tập 5.(***)

Xây dựng một bộ gán nhãn từ loại đơn giản sử dụng Hidden Markov Model(bigram HMM tagger).

- Cần bộ ngữ liệu đã gán nhãn (part-of-speech-tagged corpus)
- Chia ngữ liệu thành 2 tập: huấn luyện (training set) và tập kiểm thử (testset)
- Tính các xác suất chuyển trạng thái và xác suất quan sát (transition andobservation probabilities)
- Triển khai thuật toán Viterbi-Chạy thuật toán trên tập kiểm thử và và ghi nhận tỷ lệ lỗi (error rate), F1-scores...

```python
from sklearn.metrics import f1_score, classification_report

def train_hmm_tagger(tagged_sentences):
    transition_counts = {}
    observation_counts = {}
    tag_counts = {}

    for sentence in tagged_sentences:
        previous_tag = 'START'
        for word, tag in sentence:
            if previous_tag is not None:
                transition_counts[(previous_tag, tag)] = transition_counts.get((previous_tag, tag), 0) + 1
            observation_counts[(tag, word)] = observation_counts.get((tag, word), 0) + 1
            tag_counts[tag] = tag_counts.get(tag, 0) + 1
            previous_tag = tag

    transition_probabilities = {
        k: v / tag_counts[k[0]] for k, v in transition_counts.items()
            if k[0] != 'START'
    }
    start_total = sum(v for (p, _), v in transition_counts.items() if p == 'START')
    for (prev, tag), count in transition_counts.items():
        if prev == 'START':
            transition_probabilities[('START', tag)] = count / start_total

    observation_probabilities = {k: v / tag_counts[k[0]] for k, v in observation_counts.items()}

    return transition_probabilities, observation_probabilities, tag_counts
```

```python
def viterbi_algorithm(sentence, transition_probabilities, observation_probabilities, tag_counts):
    V = [{}]
    path = {}

    for tag in tag_counts:
        V[0][tag] = (observation_probabilities.get((tag, sentence[0]), 1e-10) *
                     transition_probabilities.get(('START', tag), 1e-10))
        path[tag] = [tag]

    for t in range(1, len(sentence)):
        V.append({})
        new_path = {}
        for tag in tag_counts:
            (prob, state) = max(
                (V[t-1][prev_tag] *
                 transition_probabilities.get((prev_tag, tag), 1e-10) *
                 observation_probabilities.get((tag, sentence[t]), 1e-10),
                 prev_tag)
                for prev_tag in tag_counts
            )
            V[t][tag] = prob
            new_path[tag] = path[state] + [tag]
        path = new_path

    n = len(sentence) - 1
    (prob, state) = max((V[n][tag], tag) for tag in tag_counts)
    return path[state]
```

```python
# Bộ part-of-speech-tagged corpus để demo
tagged_sentences = [
    [('The', 'DT'), ('quick', 'JJ'), ('brown', 'JJ'), ('fox', 'NN')],
    [('jumps', 'VBZ'), ('over', 'IN'), ('the', 'DT'), ('lazy', 'JJ'), ('dog', 'NN')]
]

# Chia dữ liệu thành tập train và test
train_data = tagged_sentences[:1]
test_data  = tagged_sentences[1:]

#Train HMM tagger
transition_probabilities, observation_probabilities, tag_counts = train_hmm_tagger(train_data)
#transition andobservation probabilities
print("Transition Probabilities:")
for (prev, tag), prob in transition_probabilities.items():
    print(f"P({tag} | {prev}) = {prob:.4f}")
print("\nObservation Probabilities:")
for (tag, word), prob in observation_probabilities.items():
    print(f"P({word} | {tag}) = {prob:.4f}")

# Đánh giá trên tập test
all_true, all_pred = [], []

for sentence in test_data:
    words     = [w for w, _ in sentence]
    true_tags = [t for _, t in sentence]
    pred_tags = viterbi_algorithm(words, transition_probabilities,
                                  observation_probabilities, tag_counts)

    print(f"Words    : {words}")
    print(f"True tags: {true_tags}")
    print(f"Pred tags: {pred_tags}")
    print()

    all_true.extend(true_tags)
    all_pred.extend(pred_tags)

print(classification_report(all_true, all_pred, zero_division=0))
```
