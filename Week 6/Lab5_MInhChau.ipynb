```python
pd.pivot_table(df, values='mpg', index='cyl', aggfunc='mean')
```

## 2. Group by cyl, then apply aggregate funcƟons (mean) on [‘mpg’, ‘hp’]

```python
pd.pivot_table(df, values=['mpg','hp'], index='cyl', aggfunc='mean')
```

## 3. Group by [cyl, am], then apply aggregate funcƟons (mean) on [‘mpg’, ‘hp’]

```python
pd.pivot_table(df, values=['mpg','hp'], index=['cyl','am'], aggfunc='mean')
```

```python
pd.pivot_table(df, values=['mpg','hp'], index='cyl', aggfunc=['mean','std'])
```

## 5. Group by [cyl, am], then apply aggregate functions (mean, std) on [‘mpg’, ‘hp’]

```python
pd.pivot_table(df, values=['mpg','hp'], index=['cyl','am'], aggfunc=['mean','std'])
```

## 6. Group by [cyl, am], then apply aggregate functions (mean, median, std) on ‘mpg’ and  then apply aggregate functions(min, max) on ‘hp’.Also, usefill_value = 0, margins = True, margins_name = “Total”.

```python
pd.pivot_table(
    df,
    values='mpg',
    index=['cyl', 'am'],
    columns='gear',
    aggfunc=['mean','std', 'median'],
    fill_value=0,
    margins=True,
    margins_name='Total'
)
```

```python
pd.pivot_table(
    df,
    values='hp',
    index=['cyl', 'am'],
    columns='gear',
    aggfunc=['min','max'],
    fill_value=0,
    margins=True,
    margins_name='Total'
)
```

## 7.Group by [cyl], set pivot table column as [gear],  then apply aggregate funcƟons (mean, std) on [‘mpg’, ‘hp’]. Also, usefill_value = 0, margins = True, margins_name = “Total”.

```python
pd.pivot_table(
    df,
    values=['mpg','hp'],
    index='cyl',
    columns='gear',
    aggfunc=['mean','std'],
    fill_value=0,
    margins=True,
    margins_name='Total'
)
```

## 8. Group by [cyl, am], set pivot table columns as [gear, carb], then apply aggregate funcƟons (mean, std)on [‘mpg’, ‘hp’]. Also, usefill_value = 0, margins = True, margins_name = “Total”.

```python
pd.pivot_table(
    df,
    values=['mpg','hp'],
    index=['cyl','am'],
    columns=['gear','carb'],
    aggfunc=['mean','std'],
    fill_value=0,
    margins=True,
    margins_name='Total'
)
```

## 9. Group by [cyl, am], set pivot table columns as [gear, carb], then apply aggregate functions (lambda funcƟon: sqrt(sum(x*x))) on [‘mpg’, ‘hp’]. Also, usefill_value = 0, margins = True, margins_name = “Total”.

```python
from math import sqrt

pd.pivot_table(
    df,
    values=['mpg','hp'],
    index=['cyl','am'],
    columns=['gear','carb'],
    aggfunc=lambda x: sqrt(sum(x*x)),
    fill_value=0,
    margins=True,
    margins_name='Total'
)
```

## 10. Use df.apply()and a lambda funcƟon, insert a new column called “Code” that uses the FIRST 3 letters of the “model” and 2 leƩers of the code of the COUNTRY: {code: countrycode}. all LOWERCASE.

```python
def get_country_code(model):
    country_dict = {
        'jp': ['toyota','honda','datsun','mazda'],
        'eu': ['merc','ferrari','volvo','porsche','maserati','lotus'],
    }
    name = model.lower()
    for code, brands in country_dict.items():
        if any(b in name for b in brands):
            return code
    return 'us'

df['country_code'] = df.apply(
    lambda x: f"{x['model'][:3].lower()}:{get_country_code(x['model'])}",
    axis=1
)
df[['model','country_code']].head(10)
```
