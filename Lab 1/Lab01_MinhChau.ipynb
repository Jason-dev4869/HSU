##### Why do we need to replace the missing data with NaN?

Replacing missing data with NaN allows us to easily identify and handle missing values in the dataset. NaN is a standard representation for missing data in pandas, and it provides various functions to handle and analyze missing values effectively. By using NaN, we can perform operations such as filtering, imputation, and analysis without encountering errors or inaccuracies caused by missing data.

##### Why '?' is replaced with NaN?

The '?' symbol is often used in datasets to indicate missing or unknown values. By replacing it with NaN (Not a Number), we can utilize the powerful data manipulation and analysis capabilities of libraries like pandas. 

NaN is a standard way to represent missing data in Python, and it allows us to perform operations such as filtering, aggregation, and statistical analysis without encountering errors that may arise from non-numeric values. 

Additionally, many functions in pandas are designed to handle NaN values appropriately, making it easier to work with incomplete datasets.

##### Why do we need to read the missing data as NaN?
 
Reading missing data as NaN is crucial for accurate data analysis and processing. NaN values allow us to identify and handle missing data effectively. When we read missing data as NaN, we can easily perform operations such as filtering out missing values, calculating statistics while ignoring NaN, and applying imputation techniques to fill in missing values. 

If we were to read missing data as a different placeholder (like '?'), it would require additional steps to identify and handle those values, which can lead to errors and complications in data analysis. 

By using NaN, we can leverage the built-in functionality of pandas and other libraries to manage missing data efficiently, ensuring that our analyses are accurate and reliable.

#### c) Count the number and percentage of rows with missing in the entire dataset 

```python
#TODO: Count the number and percent of rows with missing data in entire dataset
total_rows = len(df)
rows_with_missing_data = df.isnull().any(axis=1).sum()
percent_missing = (rows_with_missing_data / total_rows) * 100
print(f'Total rows: {total_rows}')
print(f'Rows with missing data: {rows_with_missing_data}')
print(f'Percent of rows with missing data: {percent_missing:.2f}%')
```

#### d) Count the number and percentage of missing data in each

```python
#TODO: Count the number and percentage of missing data in each rows
missing_data_per_row = df.isnull().sum(axis=1)
percent_missing_per_row = (missing_data_per_row / df.shape[1]) * 100
missing_data_summary = pd.DataFrame({
    'Missing Data Count': missing_data_per_row,
    'Missing Data Percent': percent_missing_per_row
})
missing_data_summary
```

#### e) Randomly drop 10% of the rows that have missing data. Hint: use any() andnumpy.random.choice() 

```python
#TODO: Randomly drop 10% of the rows that have missing data
rows_with_missing_data_indices = df[df.isnull().any(axis=1)].index
num_rows_to_drop = int(0.1 * len(rows_with_missing_data_indices))
rows_to_drop = np.random.choice(rows_with_missing_data_indices, size=num_rows_to_drop, replace=False)
df_dropped = df.drop(index=rows_to_drop)
df_dropped
```

#### f) Fill in the missing data with median (if the column data type is numeric) or mode (if the column data type is not numeric)

```python
#TODO: Fill the missing data with mean (if numeric) or mode (if non-numeric)
for column in df.columns:
    if df[column].dtype in [np.float64, np.int64]:  # Check if the column is numeric
        mean_value = df[column].mean()
        df[column].fillna(mean_value, inplace=True)
    else:  # Non-numeric column
        mode_value = df[column].mode()[0]
        df[column].fillna(mode_value, inplace=True)
df
```
