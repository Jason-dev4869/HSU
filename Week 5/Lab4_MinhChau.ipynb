## 1.Write Python code to create the 3 dfs above. Hint: df = pd.DataFrame(data) where data is a dicƟonary with keys = column names, values = list  of data in the corresponding columns

```python
#import libraries
import pandas as pd

#Create 3 dataframes
#Data for df_player
df_player = {
    'Player':    ['Milner', 'Countinho', 'Kane', 'Son', 'Rooney', 'Baines', 'Hazard'],
    'Team':      ['Liverpool', 'Liverpool', 'Tottenham', 'Tottenham', 'Everton', 'Everton', 'Chelsea'],
    'Goals':     [2, 5, 10, 6, 4, 1, 7],
    'Assists':   [12, 4, 4, 10, 7, 1, 8]
}
df_player = pd.DataFrame(df_player)
df_player
```

```python
#Data for df_teams
df_teams = {
    'Team':        ['Liverpool', 'Tottenham', 'Everton', 'Chelsea'],
    'Home_City':     ['Liverpool', 'London', 'Liverpool', 'London'],
    'Num_Fans':    [0.2, 2.3, 0.1, 5.1]
}
df_teams = pd.DataFrame(df_teams)
df_teams
```

```python
# Data for df_city
df_city = {
    'City':       ['London', 'Liverpool'],
    'Population': [9.0, 0.5],
}
df_city = pd.DataFrame(df_city)
df_city
```

## 2. Merge the 3 dfs (called df_merged) using only 1 line of code

```python

df_merged = pd.merge(pd.merge(df_player, df_teams, on='Team'), df_city, left_on='Home_City', right_on='City')
df_merged
```

## 3. Join the 3 dfs (called df_joined). You might need to set_index() before you could join them

```python
df_joined = df_player.set_index('Team').join(df_teams.set_index('Team')).join(df_city.set_index('City'))
df_joined
```

## 4. Remove Home_City from df_merged

```python
df_merged = df_merged.drop(columns='Home_City')
df_merged
```

## 5.Find the player with the highest goals and then print out his statistics

```python
df_merged.loc[df_merged['Goals'].idxmax()]
```

## 6.Find the total number of goals of players from the sameteams. Print out just the teams and the total goals.

```python
df_merged.groupby('Team')['Goals'].sum().reset_index()
```

## 7. Use filter, print out players from London who scored more than 2 goals.

```python
df_merged[(df_merged['City'] == 'London') & (df_merged['Goals'] > 2)]
```

## 8. Print out the TOP player (by total goals and assists) from each team. Only show his name, team, total goals and assists.

```python
df_merged['Total'] = df_merged['Goals'] + df_merged['Assists']
top_per_team = df_merged.loc[df_merged.groupby('Team')['Total'].idxmax(), ['Player', 'Team', 'Total']]
top_per_team
```

## 9. Insert a new column called “Star” that gives a “*” to players whose total goals and assists are greater than 12. Otherwise, just leave it empty.

```python
df_merged['Star'] = df_merged['Total'].apply(lambda x: '*' if x > 12 else '')
df_merged
```
