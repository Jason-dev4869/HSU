# Chapter 5: The k-Nearest Neighbors Classifiers

### Standard k-NN vs. Weighted k-NN 

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier

# -------------------------------
# 1️⃣ Tạo dữ liệu 2 lớp rõ ràng
# -------------------------------
# 3 điểm lớp 1 (màu đỏ), nằm hơi xa điểm truy vấn
X1 = np.array([[1, 1], [1, 2], [2, 1]])
y1 = np.array([1, 1, 1])

# 2 điểm lớp 2 (màu xanh), gần hơn điểm truy vấn
X2 = np.array([[3, 3], [3, 4]])
y2 = np.array([2, 2])

# Gộp dữ liệu 2 lớp
X = np.vstack((X1, X2))
y = np.concatenate((y1, y2))

# -------------------------------
# 2️⃣ Điểm cần phân loại (query)
# -------------------------------
query = np.array([[2.8, 3.2]])

# -------------------------------
# 3️⃣ Hai mô hình: uniform vs weighted
# -------------------------------
knn_uniform = KNeighborsClassifier(n_neighbors=5, weights='uniform')
knn_weighted = KNeighborsClassifier(n_neighbors=5, weights='distance')

# Huấn luyện mô hình
knn_uniform.fit(X, y)
knn_weighted.fit(X, y)

# Dự đoán
pred_uniform = knn_uniform.predict(query)[0]
pred_weighted = knn_weighted.predict(query)[0]

print("Uniform k-NN → class:", pred_uniform)
print("Weighted k-NN → class:", pred_weighted)

# -------------------------------
# 4️⃣ Trực quan hóa
# -------------------------------
plt.figure(figsize=(6,5))
plt.scatter(X1[:,0], X1[:,1], color='red', label='Class 1', s=60)
plt.scatter(X2[:,0], X2[:,1], color='blue', label='Class 2', s=60)
plt.scatter(query[0,0], query[0,1], color='yellow', edgecolor='black', s=200, marker='*', label='Query')
plt.legend()
plt.title("Uniform vs Weighted k-NN (Different Predictions)")
plt.xlabel("X1")
plt.ylabel("X2")
plt.show()
```
