## PHẦN 1: CÂU HỎI LÝ THUYẾT (Chapter 1)

Sinh viên trả lời trực tiếp vào notebook:

**1. Deep Learning là gì? Phân biệt với Machine Learning truyền thống.**

Deep Learning là một lĩnh vực con của Machine Learning, tập trung vào việc sử dụng các mạng nơ-ron sâu (deep neural networks) để học từ dữ liệu. Trong khi Machine Learning truyền thống thường dựa vào các thuật toán đơn giản và yêu cầu nhiều bước xử lý dữ liệu thủ công, Deep Learning tự động học các đặc trưng phức tạp từ dữ liệu thông qua nhiều lớp ẩn trong mạng nơ-ron. Điều này cho phép Deep Learning đạt được hiệu suất cao hơn trong các tác vụ như nhận diện hình ảnh, xử lý ngôn ngữ tự nhiên và chơi game.

**2. Neural Network hoạt động như thế nào?**

Neural Network là một mô hình học máy được lấy cảm hứng từ cấu trúc của bộ não con người. Nó bao gồm các lớp nơ-ron (neurons) được kết nối với nhau. Mỗi nơ-ron nhận đầu vào, thực hiện một phép tính (thường là một hàm kích hoạt) và truyền kết quả đến các nơ-ron ở lớp tiếp theo. Quá trình này tiếp tục qua nhiều lớp ẩn cho đến khi đạt được đầu ra cuối cùng. Neural Network học bằng cách điều chỉnh trọng số của các kết nối giữa nơ-ron dựa trên lỗi giữa dự đoán và giá trị thực tế, thường sử dụng thuật toán lan truyền ngược (backpropagation).

**3. Vai trò của dữ liệu trong deep learning.**

Dữ liệu đóng vai trò cực kỳ quan trọng trong deep learning. Mô hình deep learning cần một lượng lớn dữ liệu để học các đặc trưng phức tạp và đạt được hiệu suất tốt. Dữ liệu chất lượng cao giúp mô hình học được các mẫu và mối quan hệ trong dữ liệu, trong khi dữ liệu kém chất lượng có thể dẫn đến việc mô hình học sai hoặc không hiệu quả. Ngoài ra, việc chuẩn bị và xử lý dữ liệu (data preprocessing) cũng rất quan trọng để đảm bảo rằng dữ liệu phù hợp với mô hình và giúp cải thiện hiệu suất.

**4. Overfitting là gì? Cho ví dụ.**

Overfitting là hiện tượng khi một mô hình học quá kỹ lưỡng từ dữ liệu huấn luyện, đến mức nó không thể tổng quát hóa tốt trên dữ liệu mới (dữ liệu kiểm tra). Mô hình overfitting sẽ có hiệu suất rất cao trên dữ liệu huấn luyện nhưng lại hoạt động kém trên dữ liệu kiểm tra. Ví dụ, nếu bạn huấn luyện một mô hình để nhận diện chữ số từ một tập dữ liệu nhỏ và mô hình đó học thuộc lòng tất cả các mẫu trong tập huấn luyện, nó sẽ không thể nhận diện chính xác các chữ số mới mà nó chưa từng thấy trước đó.

## PHẦN 2: TOÁN HỌC CƠ BẢN (Chapter 2)

### Bài tập 1: Vector và ma trận
Cho vector x = [1, 2, 3] và w = [0.2, 0.5, 0.3]

Yêu cầu:
- Tính dot product
- Tính sigmoid(dot)


```python
import numpy as np

x = np.array([1, 2, 3])
w = np.array([0.2, 0.5, 0.3])

# TODO: tính dot product
dot_product = np.dot(x, w)

# TODO: tính sigmoid
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

sigmoid_value = sigmoid(dot_product)
print("Dot Product:", dot_product)
print("Sigmoid Value:", sigmoid_value)
```

### Bài tập 2: Gradient đơn giản
Cho hàm f(x) = x^2

Yêu cầu:
- Tính đạo hàm
- Viết code tính gradient tại x = 3


```python
# TODO: tính gradient
def gradient(x):
    return 2 * x

x_value = 3
grad_value = gradient(x_value)
print("Gradient at x =", x_value, "is", grad_value)
```

## PHẦN 3: LÀM QUEN KERAS & TENSORFLOW (Chapter 3)

### Bài tập 3: Tạo model đơn giản

Yêu cầu:
- Tạo Sequential model
- 2 hidden layer (ReLU) gồm 6 và 8 output neurons
- 1 output layer (sigmoid)


```python
from tensorflow import keras
from tensorflow.keras import layers

# TODO: xây dựng model
model = keras.Sequential([
    layers.Dense(6, activation='relu'),
    layers.Dense(8, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])
```

### Bài tập 4: Compile model

Yêu cầu:
- optimizer: adam
- loss: binary_crossentropy
- metrics: accuracy


```python
# TODO: compile model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

## PHẦN 4: CLASSIFICATION (Chapter 4)

### Bài tập 5: Phân loại nhị phân

Dataset: Breast Cancer Wisconsin (sklearn)

Yêu cầu:
- Load dữ liệu
- Chia train/test
- Train model
- Evaluate accuracy


```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split

# TODO: load data
data = load_breast_cancer()

# TODO: split data
x = data.data
y = data.target
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)

# TODO: train model
model.fit(x_train, y_train, epochs=10, batch_size=32)

# TODO: evaluate
loss, accuracy = model.evaluate(x_test, y_test)
print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

### Bài tập 6: Phân tích kết quả

Yêu cầu:
- In confusion matrix
- Nhận xét model:

Dưa vào confusion matrix, mô hình đã dự đoán đúng 37 trường hợp âm tính (True Negatives) và 67 trường hợp dương tính (True Positives). Tuy nhiên, có 5 trường hợp âm tính bị dự đoán sai là dương tính (False Positives) và 5 trường hợp dương tính bị dự đoán sai là âm tính (False Negatives). Điều này cho thấy mô hình có một số lỗi trong việc phân loại, nhưng tổng thể vẫn đạt được độ chính xác cao.

```python
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# TODO: tính confusion matrix
y_pred = (model.predict(x_test) > 0.5).astype("int32")
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", cm)

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')
plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix')
plt.show()
```

## PHẦN 5: REGRESSION (Chapter 4)

### Bài tập 7: Dự đoán giá nhà

Dataset: California Housing

Yêu cầu:
- Xây dựng model regression
- Loss: mse
- Metric: mae


```python
from sklearn.datasets import fetch_california_housing

# TODO: load data
data = fetch_california_housing()
X = data.data
y = data.target
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# TODO: build model regression
model = keras.Sequential([
    layers.Dense(64, activation='relu'),
    layers.Dense(64, activation='relu'),
    layers.Dense(1)
])

# TODO: train
model.compile(optimizer='adam', loss='mse')
model.fit(x_train, y_train, epochs=100, batch_size=32)
```

### Bài tập 8: Đánh giá mô hình regression

Yêu cầu:
- Tính MAE
- Vẽ biểu đồ loss


```python
import matplotlib.pyplot as plt

# TODO: plot loss
mae = model.history.history['loss']
plt.plot(mae)
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.show()
```

## PHẦN 6: BÀI TẬP MỞ RỘNG

1. Thay đổi số layer → ảnh hưởng thế nào?

Khi thay đổi số layer trong một mô hình học sâu, nó có thể ảnh hưởng đến khả năng học và hiệu suất của mô hình. Thêm nhiều layer có thể giúp mô hình học được các đặc trưng phức tạp hơn từ dữ liệu, nhưng cũng có thể dẫn đến overfitting nếu mô hình quá phức tạp so với lượng dữ liệu huấn luyện. Ngược lại, nếu số layer quá ít, mô hình có thể không đủ khả năng học được các đặc trưng quan trọng, dẫn đến underfitting.

2. Thay đổi learning rate → kết quả ra sao?

Learning rate thay đổi thì kết quả của quá trình huấn luyện sẽ bị ảnh hưởng đáng kể. Nếu learning rate quá cao, mô hình có thể không hội tụ và mất ổn định, dẫn đến việc mất mát tăng lên hoặc dao động quanh giá trị tối ưu. Nếu learning rate quá thấp, quá trình huấn luyện sẽ rất chậm và có thể mắc kẹt ở các điểm tối ưu cục bộ, dẫn đến hiệu suất kém.

3. So sánh overfitting vs underfitting

- Mô hình overfitting có độ chính xác cao trên tập huấn luyện nhưng thấp trên tập kiểm tra. Ngược lại, underfitting xảy ra khi mô hình quá đơn giản để học được các đặc trưng quan trọng từ dữ liệu, dẫn đến hiệu suất kém cả trên tập huấn luyện và tập kiểm tra. 
- Mô hình underfitting có độ chính xác thấp trên cả hai tập dữ liệu.
