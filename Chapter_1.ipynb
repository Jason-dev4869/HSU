# Chương 1: Nhập môn về Trí tuệ Nhân tạo

### 1. Trí tuệ nhân tạo (Artifacial Intelligent)

#### 1.1. Khái niệm Trí tuệ nhân tạo

- Khái niệm "Trí tuệ nhân tạo" lần đầu vào năm 1955 bởi John McCarthy (1927 - 2011)
- Trí nhân nhân tạo (AI) được hiểu là khả năng tư duy và tự học hỏi của máy móc hoặc một chương trình máy tính, nhằm thực hiện các nhiệm vụ phức tạp và có tính sáng tạo cao, tương tự như con người
- Quy trình:  
    + Học tập (thu nhận thông tin và quy tắc sử dụng thông tin)
    + Suy luận (áp dụng quy tắc để đưa ra kết luận gần đúng hoặc chính xác) 
    + Tự điều chỉnh



#### 1.2. Phân loại Trí tuệ nhân tạo (AI)

- Trí tuệ nhân tạo hẹp - Artificial Narrow Intelligence (ANI) [*Trọng tâm của khóa học SIC*]
    +  Trí tuệ nhân tạo yếu thực hiện các chức năng cụ thể của con người (VD: loa thông minh, xe tự lái)
    + Phần lớn các tiến bộ công nghệ gần đây đều đén từ lĩnh vực này
    
- Trí tuệ nhân tạo tổng quát - Artificial General Intelligence (AGI)
    + Trí tuệ nhân tạo mạnh thực hiện được mọi thao tác như con người
    + Tiến bộ của AGI hiện đang chậm hơn nhiều so với ANI
- Sơ đồ tập con Trí tuệ nhân tạo



### 2. Học Máy (Machine Learning)

#### 2.1. Khái niệm Học Máy

- Học máy được coi là một lĩnh vực trí tuệ nhân tạo và là nghiên cứu về các thuật toán máy tính có khả năng tự động cải thiện thông qua các ví dụ và kinh nghiệm.

- Chương trình học máy ấy sẽ được thiết kế để máy tính học hỏi qua trải nghiệm (E) đối với một số loại nhiệm vụ cụ thể (T) và thước đo hiệu suất (P). Với P thuộc T và cải thiện nhờ E dựa vào P.

- Example: 
    Ta có một chương trình phân loại hình ảnh chó mèo:
    + Training data: Một tập dữ liệu hình ảnh chó mèo
    + Task (T): Phân loại hình ảnh chó mèo
    + Performance metric (P): Tỷ lệ chính xác phân loại hình ảnh chó mèo
    + Experience (E): Tập dữ liệu hình ảnh chó mèo được sử dụng để đào tạo mô hình

    Để đạt được T, ta phải thiết kế cấu trúc giải thuật sao cho máy có thể học được nhưng dữ liệu có trong Training data (10000 ảnh) và đo đạt tính chính xác P của mô hình, dựa vào P ta rút ra được E để cải thiện mô hình.

- Quá trình cải thiện thông qua E có phương pháp sau:
    + Phương pháp học suy diễn (Inductive Learning): Là quá trình mà mô hình học hỏi từ dữ liệu thông qua suy luận suy diễn
    + Phương pháp học quy nạp (Deductive Learning): Là quá trình mà mô hình học hỏi từ dữ liệu giữ lại các mẫu tốt khi training
    + Phương pháp học Ockham's Razor (Occam's Razor): Là phương pháp học hỏi từ dữ liệu bằng cách chọn mô hình đơn giản nhất có thể giải thích dữ liệu tốt nhất.

#### 2.2. Các ngành liên quan tới Học Máy

- Học máy là một lĩnh vực liên ngành kết hợp nền tảng học thuật và thành tựu trong các lĩnh vực khác nhau, thay vì chỉ giới hạn ở công nghệ phương pháp luận trong bất kỳ lĩnh vực nào.
- Bao gồm: 
    + Xác suất và thống kê 
    + Khoa học máy tính
    + Lý thuyết cơ sở dữ liệu
    + Khoa học nhận thức
    + Khoa học thần kinh và nhận dạng mẫu



##### 2.2.1. Học Máy vs Thống kê dữ liệu

- Thống kê được cho là nền tảng lý thuyết cung cấp một phương pháp khoa học và có hệ thống để chuyển đổi dữ liệu thành thông tin nhằm giải thích mức độ phù hợp của dữ liệu đã cho với giả thuyết và lý thuyết của nhà nghiên cứu

- Học Máy được sử dụng để giải quyết các nhiệm vụ khó thiết kế hoặc khó lập trình các thuật toán rõ ràng nhằm định lượng mối quan hệ phức tạp bằng cách xác định đặc trưng của các cơ chế tiềm năng do dữ liệu tạo ra

- Thoạt nhìn dù đối lặp, tuy nhiên trên thực tế nhiều phương pháp học máy dựa trên việc học thống kê từ các phương pháp thống kê
