### 1. Specify formally:

```python
# Initial state (s_0)
# The initial state of the game is an empty 3x3 grid, 
# represented as a list of lists.
s_0 = [['', '', ''],
       ['', '', ''],
       ['', '', '']]
```

```python
# Actions(s)
# The actions available in a given state are the empty spaces on the grid where a player can place their mark (X or O).
def actions(s):
    return [(i, j) for i in range(3) for j in range(3) if s[i][j] == '']
```

```python
# Result(s, a)
# The result of taking an action (placing a mark on the grid) is a new state of the game with that mark in place.
def result(s, a, player):
    new_state = [row[:] for row in s]  # Deep copy of the state
    new_state[a[0]][a[1]] = player
    return new_state
```

```python
# Terminal(s)
# A state is terminal if there is a winner or if there are no more empty spaces (a draw).
def terminal(s):
    # Check rows and columns
    for i in range(3):
        if s[i][0] == s[i][1] == s[i][2] != '':
            return True
        if s[0][i] == s[1][i] == s[2][i] != '':
            return True
    # Check diagonals
    if s[0][0] == s[1][1] == s[2][2] != '':
        return True
    if s[0][2] == s[1][1] == s[2][0] != '':
        return True
    # Check for draw (no empty spaces)
    if all(s[i][j] != '' for i in range(3) for j in range(3)):
        return True
    return False
```

```python
# Utility(s)
# The utility of a terminal state is 1 if the player wins, -1 if the opponent wins, and 0 for a draw or non-terminal state.
def utility(s, player):
    opponent = 'O' if player == 'X' else 'X'
    # Check rows and columns
    for i in range(3):
        if s[i][0] == s[i][1] == s[i][2] == player:
            return 1
        if s[0][i] == s[1][i] == s[2][i] == player:
            return 1
    # Check diagonals
    if s[0][0] == s[1][1] == s[2][2] == player:
        return 1
    if s[0][2] == s[1][1] == s[2][0] == player:
        return 1
    # Check for draw
    if all(s[i][j] != '' for i in range(3) for j in range(3)):
        return 0
    # If opponent wins, return -1
    for i in range(3):
        if s[i][0] == s[i][1] == s[i][2] == opponent:
            return -1
        if s[0][i] == s[1][i] == s[2][i] == opponent:
            return -1
    if s[0][0] == s[1][1] == s[2][2] == opponent:
        return -1
    if s[0][2] == s[1][1] == s[2][0] == opponent:
        return -1
    return 0
```

### 2. Explain why Tic-Tac-Toe is considered:
   - A zero-sum game: The gain of one player is exactly the loss of the other player. 
   If one player wins, the other player loses, and if it's a draw, neither player gains or loses anything.  
   
   - A deterministic game: The outcome of the game is determined solely by the players' actions, with no element of chance involved. 
   There are no random events or hidden information that can affect the outcome.  
   
   - A game with perfect information: Both players have complete knowledge of the game state at all times. 
   There are no hidden cards, dice rolls, or secret information that can influence the players' decisions. 
   Each player can see the entire board and all of the opponent's moves, 
   allowing them to make informed decisions based on the current state of the game.
