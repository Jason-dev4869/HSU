## 3. BFS & DFS Visualization (25 minutes)

### Exercise 7: BFS Traversal
Implement BFS traversal on a graph.

Requirements:
1. Use a graph with at least 6 nodes
2. Track visited nodes step-by-step
3. Visualize each step using Graphviz
4. Highlight visited nodes using colors

```python
#Implement BFS traversal on a graph.
#Requirements:
#1. Use a graph with at least 6 nodes
#2. Track visited nodes step-by-step
#3. Visualize each step using Graphviz
#4. Highlight visited nodes using colors
from collections import deque
from IPython.display import display
bfs_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

def draw_bfs_step(visited):
    dot = Digraph(node_attr={'shape': 'circle'})
    for node in bfs_graph:
        color = 'lightblue' if node in visited else 'white'
        dot.node(node, style='filled', fillcolor=color)
    for u in bfs_graph:
        for v in bfs_graph[u]:
            dot.edge(u, v)
    return dot

def bfs(start):
    visited = [start]
    queue = deque([start])
    steps = [draw_bfs_step(visited.copy())]

    while queue:
        node = queue.popleft()
        for neighbor in bfs_graph[node]:
            if neighbor not in visited:
                visited.append(neighbor)
                steps.append(draw_bfs_step(visited.copy()))
                queue.append(neighbor)

    return steps

bfs_steps = bfs('A')
for i, step in enumerate(bfs_steps):
    print(f"Step {i+1}")
    display(step)
```

### Exercise 8: DFS Traversal
Implement DFS traversal.

Requirements:
1. Use the same graph as BFS
2. Visualize traversal step-by-step
3. Use a different color scheme from BFS

```python
dfs_graph = bfs_graph  # Reuse the same graph
def draw_dfs_step(visited):
    dot = Digraph(node_attr={'shape': 'circle'})
    for node in dfs_graph:
        color = 'lightgreen' if node in visited else 'white'
        dot.node(node, style='filled', fillcolor=color)
    for u in dfs_graph:
        for v in dfs_graph[u]:
            dot.edge(u, v)
    return dot
def dfs(start):
    visited = []
    stack = [start]
    steps = [draw_dfs_step(visited.copy())]

    while stack:
        node = stack.pop()
        if node not in visited:
            visited.append(node)
            steps.append(draw_dfs_step(visited.copy()))
            for neighbor in reversed(dfs_graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)

    return steps
dfs_steps = dfs('A')
for i, step in enumerate(dfs_steps):
    print(f"Step {i+1}")
    display(step)
```

### Exercise 9: Animation
Create an animation of BFS or DFS.

Requirements:
1. Generate multiple frames for each step
2. Display animation inside notebook OR export as GIF
3. Control animation speed

```python
#BFS animation
from IPython.display import clear_output
import time
import imageio.v2 as imageio
import os
def animate(steps, delay=1):
    for step in steps:
        clear_output(wait=True)  # xóa frame trước
        display(step)            # hiển thị frame mới
        time.sleep(delay)        # dừng 1 chút

def export_gif(frames, filename="animation.gif", duration=1):
    images = []

    for i, frame in enumerate(frames):
        filepath = f"temp_frame_{i}"

        # render ra PNG
        frame.render(filepath, format='png', cleanup=True)

        # đọc ảnh
        img = imageio.imread(f"{filepath}.png")
        images.append(img)

        # xóa file tạm
        os.remove(f"{filepath}.png")

    # tạo GIF
    imageio.mimsave(filename, images, duration=duration)
```

```python
animate(bfs_steps, delay=1)
export_gif(bfs_steps, filename="bfs_animation.gif", duration= 0.5)
```

```python
animate(dfs_steps, delay=1)
export_gif(dfs_steps, filename="dfs_animation.gif", duration=1)
```

### Exercise 10: Comparison
Compare BFS and DFS.

#### Requirements:
##### 1. Show traversal order for both algorithms
BFS: A → B → C → D → E → F

DFS: A → B → D → E → F → C

##### 2. Explain differences in behavior
BFS dùng queue (FIFO) — khám phá hết tất cả hàng xóm cùng level rồi mới đi sâu hơn. Nên nó luôn tìm được đường ngắn nhất tính theo số cạnh.

DFS dùng stack (LIFO) — chọn một hướng rồi đâm thẳng xuống hết mức có thể, gặp ngõ cụt mới quay lại. Tốn ít bộ nhớ hơn trên đồ thị rộng, phù hợp cho phát hiện chu trình và backtracking.

#### 3. Provide visual comparison
