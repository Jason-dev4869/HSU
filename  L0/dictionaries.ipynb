```python
capitals = {'California': 'Sacramento', 'Texas': 'Austin', 'Colorado': 'Denver'}
```

```python
print(capitals)
```

```python
print(capitals["California"])
```

```python
print(capitals["Florida"])
```

```python
print(capitals.get("Florida","Unknown"))
```

```python
capitals["Colorado"] = "Denver" # Add a new key-value pair
print(capitals) # Output: {'California': 'Sacramento', 'Texas': 'Austin', 'Colorado': 'Denver'}
capitals["Texas"] = "Houston" # Update an existing value
print(capitals) # Output: {'California': 'Sacramento', 'Texas': 'Houston', 'Colorado': 'Denver'}
```

```python
del capitals["Texas"]
print(capitals) # Output: {'California': 'Sacramento', 'Colorado': 'Denver'}
```

```python
states = {
    "Texas": {"capital": "Austin", "flower": "Bluebonnet"},
    "California": {"capital": "Sacramento", "flower": "California Poppy"}
}
print(states["Texas"]["flower"]) # Output: Bluebonnet
```

```python
for state_name in capitals:
    print(state_name) # Output: California, Colorado
```

```python
for capital in capitals.values():
    print(capital)
```

```python
for state, capital in capitals.items():
    print(f"The capital of {state} is {capital}")
```
