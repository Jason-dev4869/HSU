```python
even_numbers = [x for x in range(10) if x % 2 == 0] 
print(even_numbers)
```

```python
my_list = ['a', 'b', 'c', 'd']
my_list[1] = 'x'
print(my_list) # Output: ['a', 'x', 'c', 'd']
my_list[1:3] = ['y', 'z'] # Replace a slice
print(my_list) # Output: ['a', 'y', 'z', 'd']
```

```python
numbers = [5, 3, 4, 1]
numbers.sort() # Output: [1, 3, 4, 5]
colors = ["red", "blue", "green", "yellow"]
colors.sort(key=len) # Sorts by string length: ['red', 'blue', 'green', 'yellow']
print(numbers)
print(colors)
```

```python
print(sum(numbers))
print(min(numbers))
print(max(numbers))
```
