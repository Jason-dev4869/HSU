## Q1. Clustering with k-means

### Task 1
Generate a 2D synthetic dataset with three clusters.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
```

```python
X, y = make_blobs(n_samples=300, centers=3, random_state=42)

#red, green, blue
colors = ['red', 'green', 'blue']
plt.figure(figsize=(8, 6))
for i in range(3):
    plt.scatter(X[y == i][:, 0], X[y == i][:, 1], color=colors[i], label=f'Cluster {i}')
plt.title('Generated Blobs')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

### Task 2
Apply the k-means algorithm with k = 3.

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)
labels = kmeans.labels_
centers = kmeans.cluster_centers_
```

### Task 3
Visualize the clustering result:
-	Color data points according to their cluster
-	Mark cluster centers clearly
-	Add a legend for clusters

Use the following color convention:
-	Cluster 0: red
-	Cluster 1: green
-	Cluster 2: blue

```python
#red, green, blue
colors = ['red', 'green', 'blue']
plt.figure(figsize=(8, 6))
for i in range(3):
    plt.scatter(X[labels == i][:, 0], X[labels == i][:, 1], color=colors[i], label=f'Cluster {i}')
plt.scatter(centers[:, 0], centers[:, 1], color='gold', marker='X', s=200, label='Centroids', edgecolor='black')
plt.title('K-Means Clustering')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

## Q2. Predicting Clusters for New Data

### Task 4
Generate 5 new random data points.

```python
#Generate new 5 point
X_new = np.random.rand(5, 2) * 10
print("New data points:\n", X_new)
```

### Task 5
Predict the cluster label for each new point using the trained k-means model.

```python
labels_new = kmeans.predict(X_new)
print("Predicted cluster labels for new data points:", labels_new)
```

### Task 6
Visualize:
-	Original clustered data
-	New data points using star markers
-	Colors of new points must match the predicted cluster colors

```python
plt.figure(figsize=(8, 6))
for i in range(3):
    plt.scatter(X[labels == i][:, 0], X[labels == i][:, 1], color=colors[i], label=f'Cluster {i}')
plt.scatter(centers[:, 0], centers[:, 1], color='gold', marker='X', s=200, label='Centroids', edgecolor='black')
for i in range(len(X_new)):
    plt.scatter(X_new[i, 0], X_new[i, 1], color=colors[labels_new[i]], marker='*', s=200, label=f'New Point {i+1}' if i == 0 else "")
plt.title('K-Means Clustering with New Data Points')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.legend()
plt.show()
```

**2.	Why is feature scaling important for distance-based algorithms?**

Feature scaling is important for distance-based algorithms because it ensures that all features contribute equally to the distance calculations. If features are on different scales, the algorithm may give more weight to those with larger ranges, leading to biased results.

**3.	What is the role of PCA in data visualization?**

PCA (Principal Component Analysis) is a dimensionality reduction technique that transforms high-dimensional data into a lower-dimensional space while preserving as much variance as possible. In data visualization, PCA helps to visualize complex datasets by reducing them to 2 or 3 dimensions, making it easier to identify patterns and relationships between data points.

**4.	How does k-means assign a data point to a cluster?**

K-means assigns a data point to a cluster by calculating the distance between the point and each cluster centroid, then assigning the point to the cluster with the nearest centroid.

**5.	What are the limitations of k-means clustering?**
    
The limitations of k-means clustering include:
- It assumes that clusters are spherical and equally sized, which may not always be the case.
- It is sensitive to outliers, which can skew the centroids and lead to poor clustering results.
- It requires the number of clusters (k) to be specified in advance, which can be difficult to determine.

**6.  What is the functionality of the StandardScaler class?**

The StandardScaler class in scikit-learn is used for feature scaling. It standardizes features by removing the mean and scaling to unit variance. This means that it transforms the data such that the distribution of each feature has a mean of 0 and a standard deviation of 1, which is important for many machine learning algorithms that are sensitive to the scale of the data.
