### 1. Khởi tạo dữ liệu
Pandas cho phép **tạo Series và DataFrame** từ nhiều cấu trúc dữ liệu cơ bản như List, Dict, hoặc Numpy array.

```python
import pandas as pd

# Khởi tạo DataFrame từ Dict (Ví dụ:)
data = {'Name': ['Mai', 'Tuan', 'Lan'], 'Age': [20, 22, 19] }
df = pd.DataFrame(data)
print("DataFrame đã khởi tạo:")
print(df)
```

### 2. Đọc và ghi dữ liệu
Các hàm như **`pd.read_csv()`** được sử dụng để đọc dữ liệu từ các tập tin phổ biến như CSV, Excel, hoặc JSON. Ngược lại, **`pd.to_csv()`** dùng để ghi dữ liệu ra file.

```python
# Ví dụ hàm đọc/ghi dữ liệu (Conceptual usage)

print("Sử dụng pd.read_csv()/pd.to_csv() để thao tác với file.")

# Ghi DataFrame hiện tại ra file CSV
df.to_csv("my_data.csv", index=False)

# Đọc file CSV
df_csv = pd.read_csv("my_data.csv")

# In df_csv
print(df_csv)
```

### 3. Khám phá dữ liệu
Các phương thức **`df.head()`, `df.tail()`** giúp xem nhanh vài dòng đầu hoặc cuối. **`df.info()`** cung cấp thông tin chung về kiểu dữ liệu và số lượng giá trị không thiếu, trong khi **`df.describe()`** cung cấp thống kê nhanh.

```python
# Xem 3 dòng đầu tiên (df được tạo ở bước 1)
print("3 dòng đầu tiên (df.head()):")
print(df.head(3))

# Xem thông tin chung
print("\nThông tin chung (df.info()):")
df.info()
```

### 4. Truy cập và chọn dữ liệu
Có thể **truy cập cột** bằng cách dùng nhãn cột (ví dụ: `df["col"]`). Đối với truy cập theo dòng hoặc vị trí, ta dùng **`df.iloc` (theo index số)** hoặc **`df.loc` (theo nhãn)**.

```python
# Truy cập cột 'Name'
names = df["Name"]
print("Truy cập cột 'Name':")
print(names)

# Truy cập dòng đầu tiên theo index số
first_row_iloc = df.iloc
print("\nTruy cập dòng đầu tiên (iloc):")
print(first_row_iloc)

# Cắt lát dữ liệu (lấy dòng 0 đến 1, cột 0 đến 1)
slice_data = df.iloc[0:2, 0:2]
print("\nCắt lát dữ liệu (iloc[0:2, 0:2]):")
print(slice_data)
```

### 5. Lọc dữ liệu (Filtering)
**Lọc dữ liệu** được thực hiện bằng cách áp dụng các điều kiện logic, ví dụ: `df[df["Age"] > 30]`. Ta có thể **kết hợp nhiều điều kiện** bằng toán tử `&` (AND) hoặc `|` (OR).

```python
# Lọc các dòng có 'Age' lớn hơn 25
df_filtered_age = df[df["Age"] > 25]
print("Dòng có Age > 25:")
print(df_filtered_age)

# Kết hợp điều kiện (ví dụ: Age > 20 VÀ Name là 'Tuan')
df_combined_filter = df[(df["Age"] > 20) & (df["Name"] == "Tuan")]
print("\nDòng thỏa mãn điều kiện kết hợp:")
print(df_combined_filter)
```

### 6. Thêm và xóa dữ liệu
Sinh viên có thể **thêm cột mới** bằng cách gán giá trị trực tiếp (`df["new_col"] = ...`). **Xóa cột** sử dụng phương thức **`df.drop("col", axis=1)`** và **xóa dòng** bằng cách chỉ định index (`df.drop(index)`).

```python
# Thêm cột 'City'
df["City"] = ["Hanoi", "HCMC", "Hanoi"]
print("DataFrame sau khi thêm cột City:")
print(df)

# Xóa cột 'City' (axis=1 là xóa cột)
df = df.drop("City", axis=1)
print("\nDataFrame sau khi xóa cột City:")
print(df)
```

### 7. Xử lý dữ liệu thiếu
Sử dụng **`df.isnull().sum()`** để kiểm tra số lượng giá trị thiếu trong mỗi cột. Ta có thể **điền giá trị thay thế** bằng **`df.fillna(value)`** hoặc **xóa các dòng có dữ liệu thiếu** bằng **`df.dropna()`**.

```python
import numpy as np

# Tạo DataFrame có giá trị thiếu
df_missing = pd.DataFrame({'Score': [10, np.nan, 8], 'Pass': [True, False, np.nan]})
print(df_missing)

# Kiểm tra giá trị thiếu
print("Số lượng giá trị thiếu trong mỗi cột:")
print(df_missing.isnull().sum())

# Điền giá trị 0 vào chỗ thiếu trong cột 'Score'
df_filled = df_missing.fillna({'Score': 0})
print("\nSau khi điền giá trị thiếu trong cột 'Score' (fillna):")
print(df_filled)
```

### 8. Thống kê và tổng hợp
Các hàm thống kê cơ bản bao gồm **`df.mean()`, `df.sum()`, `df.max()`**. Để thực hiện **tổng hợp theo nhóm**, sử dụng phương thức **`df.groupby("col").mean()`**.

```python
# Tính tuổi trung bình (sử dụng df gốc)
avg_age = df["Age"].mean()
print(f"Tuổi trung bình: {avg_age:.2f}")

# Đếm số lần xuất hiện của các giá trị trong cột Age
counts = df["Age"].value_counts()
print("\nSố lần xuất hiện của các giá trị Age (value_counts):")
print(counts)

# Tạo DataFrame mới để thử groupby
df_group = pd.DataFrame({'City': ['Hanoi', 'HCMC', 'Hanoi'], 'Salary': [2000, 2500, 2400]})

# Gộp nhóm theo City và tính trung bình lương
group_avg = df_group.groupby("City")["Salary"].mean()
print("\nLương trung bình theo thành phố (groupby):")
print(group_avg)
```

### 9. Sắp xếp dữ liệu
Dữ liệu được **sắp xếp** bằng hàm **`df.sort_values("col")`**. Có thể sắp xếp theo nhiều cột và chỉ định thứ tự (tăng dần/giảm dần).

```python
# Sắp xếp df gốc theo cột 'Age' giảm dần
df_sorted = df.sort_values("Age", ascending=False)
print("Sắp xếp theo Age (giảm dần):")
print(df_sorted)

# Ví dụ sắp xếp theo nhiều cột
df_sort_multi = pd.DataFrame({'Group': ['A', 'B', 'A', 'B'], 'Value': [10, 20, 8, 16]})
df_sorted_multi = df_sort_multi.sort_values(["Group", "Value"], ascending=[True, False])
print("\nSắp xếp theo Group (tăng) và Value (giảm):")
print(df_sorted_multi)
```

### 10. Kết hợp dữ liệu
Để **nối DataFrame** theo hàng hoặc cột, dùng **`pd.concat([df1, df2])`**. Để **gộp DataFrame** dựa trên khóa chung, sử dụng **`pd.merge(df1, df2, on="col")`**.

```python
df_left = pd.DataFrame({'ID': [1, 2], 'Data1': ['A', 'B']})
df_right = pd.DataFrame({'ID': [1, 2], 'Data2': ['C', 'D']})

# Nối hai DataFrame theo hàng
df_concat = pd.concat([df_left, df_right], ignore_index=True)
print("Kết hợp theo hàng (pd.concat):")
print(df_concat)

# Gộp hai DataFrame theo cột khóa 'ID' (pd.merge)
df_merged = pd.merge(df_left, df_right, on="ID", how='inner')
print("\nKết hợp theo khóa chung (pd.merge):")
print(df_merged)
```
