# Clustering with k-means

```python
# Import thư viện
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# 1. Tạo dữ liệu mẫu (2D để dễ trực quan)
X, y_true = make_blobs(n_samples=300, centers=3, cluster_std=0.60, random_state=0)

# 2. Khởi tạo mô hình KMeans với 3 cụm
kmeans = KMeans(n_clusters=3, random_state=0)

# 3. Huấn luyện mô hình
kmeans.fit(X)

# 4. Dự đoán cụm cho từng điểm
y_kmeans = kmeans.predict(X)

# 5. Vẽ kết quả phân cụm
plt.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=30, cmap='viridis')

# Vẽ tâm cụm
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.75, marker="X")
plt.title("K-means Clustering Example")
plt.show()
```

## 📝 Giải thích

- make_blobs() → tạo dữ liệu giả lập có 3 cụm, thuận tiện để minh họa.

- KMeans(n_clusters=3) → khai báo số cụm cần tìm (k=3).

- fit(X) → huấn luyện thuật toán K-means trên dữ liệu.

- predict(X) → gán mỗi điểm dữ liệu vào cụm gần nhất.

Biểu đồ → các điểm dữ liệu được tô màu theo cụm, tâm cụm (centroids) đánh dấu bằng dấu X đỏ.

## 👉 K-means khác với classification ở chỗ:

- Classification cần nhãn (supervised learning).

- Clustering không cần nhãn (unsupervised learning).
