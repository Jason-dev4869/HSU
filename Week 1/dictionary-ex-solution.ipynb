````python
# Setup: Use this dictionary for Q2 onwards unless otherwise specified.
car_dict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
````

## Q1: Dictionary Creation and Basic Access

Define a new dictionary named `inventory` representing a fruit, containing three key-value pairs: 'name', 'color', and 'price'.

Then, access and print the value associated with the 'name' key using both **bracket notation** (`[]`) and the **`.get()` method** [1].

````python
# Define the dictionary
inventory = {
    "name": "Apple",
    "color": "Red",
    "price": 0.99
}

# Access using bracket notation (throws KeyError if key doesn't exist)
print(f"Accessing using []: {inventory['name']}")

# Access using .get() method (returns None if key doesn't exist, safer)
print(f"Accessing using .get(): {inventory.get('name')}")
````

## Q2: Accessing Keys, Values, and Items (Views)

Using `car_dict`, print all keys, all values, and all key-value pairs (items) separately using the appropriate view methods [1].

````python
# Print all keys
print("Keys:", car_dict.keys())

# Print all values
print("Values:", car_dict.values())

# Print all items (key-value pairs)
print("Items:", car_dict.items())
````

## Q3: Adding and Updating Elements

**a) Adding:** Add a new key-value pair, "color": "red", to `car_dict`.

**b) Updating:** Update the 'year' of `car_dict` to 2024.

Print the updated `car_dict` after both operations [2].

````python
# a) Adding a new key (if key exists, it updates)
car_dict["color"] = "red"

# b) Updating an existing key
car_dict["year"] = 2024

print(car_dict)
````

## Q4: Deleting Elements

**a) Deletion by Key:** Use the **`.pop()` method** to remove the 'model' key from `car_dict` [2].

**b) Deletion using `del`:** Use the **`del` keyword** to remove the 'year' key from `car_dict` [2].

Print `car_dict` after each removal.

````python
# The dictionary currently is: {'brand': 'Ford', 'model': 'Mustang', 'year': 2024, 'color': 'red'}

# a) Use pop(key) to remove 'model'
car_dict.pop("model")
print("After pop('model'):", car_dict)

# b) Use del keyword to remove 'year'
del car_dict["year"]
print("After del car_dict['year']:", car_dict)
````

## Q5: Checking for Key Existence

Write an `if` statement to check if the key "brand" exists in `car_dict`. Print a confirmation message if it exists [3].

````python
# Check for key existence using the 'in' keyword
if "brand" in car_dict:
    print("Yes, 'brand' is a key in car_dict")
````

## Q6: Iteration (Looping)

Iterate through the current `car_dict` (which now only contains 'brand' and 'color') and print each key followed by its corresponding value, separated by a colon (e.g., "key: value") [2].

````python
# Iterate using the .items() method to get both key (k) and value (v)
for k, v in car_dict.items():
    print(f"{k}: {v}")
````

## Q7: `update()` Method

Create a new dictionary `specs = {"engine": "V8", "horsepower": 450}`. Use the **`.update()` method** to merge `specs` into `car_dict`. Print `car_dict` [3].

````python
specs = {"engine": "V8", "horsepower": 450}

# update() merges the specified dictionary (specs) into the current dictionary (car_dict)
car_dict.update(specs)

print(car_dict)
````

## Q8: `setdefault()` Method

Use the **`.setdefault()` method** on `car_dict` to check for the key 'mileage'. If it exists, return its value. If it does not exist, add it with a default value of 10000. Print `car_dict` [3].

````python
# 'mileage' does not exist, so setdefault adds it with the default value (10000).
car_dict.setdefault("mileage", 10000)

# If we try setdefault with an existing key (e.g., 'brand'), it just returns the current value and makes no changes.
car_dict.setdefault("brand", "UNKNOWN")

print(car_dict)
````

## Q9: `fromkeys()` Method

Create a new list of keys: `employees = ["Alice", "Bob", "Charlie"]`. Create a dictionary `employee_status` from this list using the **`.fromkeys()` method**, setting the default value for all status to "Active". Print the new dictionary [3].

````python
employees = ["Alice", "Bob", "Charlie"]

# dict.fromkeys(keys_list, default_value) creates a new dictionary
employee_status = dict.fromkeys(employees, "Active")

print(employee_status)
````

## Q10: Nested Dictionaries

Given the nested dictionary structure below, access and print the 'year' of 'child2' [3].

```python
myfamily = {
"child1": {"name": "Emil", "year": 2004},
"child2": {"name": "Tobias", "year": 2007}
}
```

````python
myfamily = {
    "child1": {"name": "Emil", "year": 2004},
    "child2": {"name": "Tobias", "year": 2007}
}

# Accessing nested data requires sequential key lookup:
# 1. Access the inner dictionary ('child2')
# 2. Access the value within that inner dictionary ('year')
child2_year = myfamily["child2"]["year"]

print(f"Year of child 2: {child2_year}")
````
