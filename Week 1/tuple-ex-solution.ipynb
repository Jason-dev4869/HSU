# Python Tuple Review Exercises - Solutions

## Exercise 1: Creation, Access, and Length

```python
# 1. Create a tuple containing mixed data types and a one-item tuple
# Tuples allow mixed data types [3]. The one-item tuple needs a trailing comma (",") [2].
data_tuple = ("Python", 3, 3.14, ("single",))
print(f"Created tuple: {data_tuple}")

# 2. Print the length of data_tuple
print(f"Length of tuple: {len(data_tuple)}") # Uses len() [3]

# 3. Access the first element (index 0) and the last element (index -1) [3]
first_element = data_tuple
last_element = data_tuple[-1]
print(f"First element: {first_element}")
print(f"Last element: {last_element}")

# 4. Use slicing from the second element (index 1) up to the last element (index -1) [3]
sliced_elements = data_tuple[1:-1]
print(f"Sliced elements (index 1 up to -1): {sliced_elements}")
```

## Exercise 2: Handling Immutability

```python
fixed_items = ("alpha", "beta", "gamma")

# 1. Explain why direct modification is not possible
# Tuples are immutable, meaning their elements cannot be changed directly after creation [2, 3].
# Example of an impossible action: fixed_items[3] = "delta"

# 2. Convert tuple to list, modify, and convert back [2]

# Step A: Convert tuple to list
temp_list = list(fixed_items)

# Step B: Modify the list (List is mutable)
temp_list[3] = "delta"

# Step C: Convert back to tuple
modified_tuple = tuple(temp_list)
print(f"Original tuple: {fixed_items}")
print(f"Final modified tuple: {modified_tuple}")
```

## Exercise 3: Concatenation and Repetition

```python
t1 = (1, 2)
t2 = (3, 4)

# 1. Concatenate (add) t1 and t2 using the '+' operator [2]
t_combined = t1 + t2
print(f"Combined tuple: {t_combined}")

# 2. Repeat t1 three times using the '*' operator [2]
t_repeated = t1 * 3
print(f"Repeated tuple: {t_repeated}")
```

## Exercise 4: Packing and Unpacking

```python
# 1. Packing: Assigning multiple values into a tuple [4]
colors = "Red", "Green", "Blue", "Yellow" # Parentheses are optional for packing
print(f"Packed tuple: {colors}")

# 2. Unpacking: Assigning tuple values to variables.
# We use the asterisk (*) before a variable to collect the remaining values into a list [4].
(primary, *secondary_mix) = colors

print(f"Primary color (first element): {primary}")
print(f"Secondary colors mix (remaining elements as a list): {secondary_mix}")
```

## Exercise 5: Iteration and Membership

```python
numbers = (10, 20, 30, 40, 50)

# 1. Iterate through the tuple using a for loop [2]
print("--- Iterating through elements ---")
for x in numbers:
    print(x)

# 2. Check for membership using the 'in' keyword [2]
print("\n--- Membership check ---")
if 30 in numbers:
    print("30 is present in the numbers tuple.")
else:
    print("30 is not present.")
```

## Exercise 6: Using Tuple Methods

```python
items = ("A", "B", "C", "A", "D", "A")

# 1. Use the count() method to determine the number of occurrences of "A" [2]
count_A = items.count("A")
print(f"Count of 'A': {count_A}")

# 2. Use the index() method to find the position of the first occurrence of "C" [4]
index_C = items.index("C")
print(f"Index of the first occurrence of 'C': {index_C}")
```
