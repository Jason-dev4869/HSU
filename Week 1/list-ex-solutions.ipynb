**Exercise 1: List Basics and Slicing.**
A List is defined as an ordered, changeable data type that allows duplicate members [1].
Create a list named `fruits` containing "apple", "banana", "cherry", "date", "elderberry". Print the **length** of the list (using `len()`). Then, use **slicing** to print the elements from the second item up to (but not including) the fourth item (`mylist[1:4]`) [1].

```python
# Khai báo list [1]
fruits = ["apple", "banana", "cherry", "date", "elderberry"]

# In độ dài list [1]
print(f"Length of list: {len(fruits)}")

# Cắt list từ index 1 (banana) đến index 3 (date, không bao gồm elderberry) [1]
print(f"Slice [1:4]: {fruits[1:4]}")
```

**Exercise 2: Access and Modification.**
Access and print the last element of the `fruits` list using **negative indexing** (`mylist[-1]`). Then, change the value of the third element ("cherry") to "grape" (`mylist[1] = "new"`). Print the updated list [1].

```python
fruits = ["apple", "banana", "cherry", "date", "elderberry"]

# Truy cập phần tử cuối cùng bằng index âm [1]
print(f"Last element: {fruits[-1]}")

# Phần tử thứ ba (index 2, vì index bắt đầu từ 0) là "cherry". Thay đổi giá trị [1]
fruits[2] = "grape"

# In list sau khi thay đổi [1]
print(f"Updated list: {fruits}")
```

**Exercise 3: Adding Elements.**
Start with an empty list called `my_list`. Use `append(item)` to add the number 10 to the end. Use `insert(index, item)` to add the number 5 at index 0. Finally, use `extend(list2)` to join the list `` to `my_list`. Print the final list [2].

```python
my_list = []

# Thêm 10 vào cuối list [2]
my_list.append(10)

# Thêm 5 vào vị trí đầu tiên (index 0) [2]
my_list.insert(0, 5)

# Nối thêm list  vào list hiện tại [2]
my_list.extend()

print(f"Final list after adding: {my_list}")
```

**Exercise 4: Deleting Elements.**
Given `numbers = [1, 2, 2, 3]`. Use `remove(item)` to delete the value 2 (first occurrence). Use `pop()` to delete the last element (since the index is not specified). Use `del mylist[index]` to delete the element at index 0. Print the resulting list [2].

```python
numbers = [1, 2, 2, 3]

# Xóa giá trị 2 đầu tiên tìm thấy [2]
numbers.remove(2)
print(f"After remove(2): {numbers}")

# Xóa phần tử cuối cùng (pop không đối số) [2]
numbers.pop()
print(f"After pop(): {numbers}")

# Xóa phần tử tại index 0 (là số 1) [2]
del numbers
print(f"After del numbers: {numbers}")
```

**Exercise 5: Iteration.**
Iterate through the list `colors = ["red", "green", "blue"]` using a `for` loop that loops over the elements, and print each color [2].

**Exercise 6: Operators and Membership.**
Given `list_A = [1, 2]` and `list_B = [3]`. Create `list_C` by concatenating `list_A + list_B`. Create `list_D` by repeating `list_A` three times (`list1 * 3`). Check if the number 4 is present in `list_C` using the `in` operator. Print `list_C`, `list_D`, and the membership result [2].

```python
list_A = [1, 2]
list_B = [3]

# Nối list (list1 + list2) [2]
list_C = list_A + list_B
print(f"Concatenation (list_C): {list_C}")

# Nhân list (list1 * 3) [2]
list_D = list_A * 3
print(f"Repetition (list_D): {list_D}")

# Kiểm tra thành viên bằng toán tử 'in' [2]
is_four_in_C = 4 in list_C
print(f"Is 4 in list_C?: {is_four_in_C}")
```

**Exercise 8: Useful Methods.**
Given `items = ['a', 'b', 'a', 'c', 'd', 'a']`. Use `count(value)` to find how many times 'a' appears. Use `index(value)` to find the index of the first occurrence of 'c'. Print the results [3].

```python
items = ['a', 'b', 'a', 'c', 'd', 'a']

# Đếm số lần xuất hiện của 'a' [3]
count_a = items.count('a')

# Tìm vị trí đầu tiên của 'c' [3]
index_c = items.index('c')

print(f"'a' appears {count_a} times.")
print(f"'c' is first found at index: {index_c}")
```

**Exercise 9: List Comprehension.**
Use **List Comprehension** to create a new list containing all even numbers between 0 and 9 (inclusive). The syntax for List Comprehension is `[x for x in range(10) if x % 2 == 0]` [3]. Print the resulting list.

```python
# Sử dụng List Comprehension để tạo list nhanh [3]
# Lấy x nếu x trong khoảng 0-9 và x là số chẵn (x % 2 == 0)
even_numbers = [x for x in range(10) if x % 2 == 0]

print(f"Even numbers (List Comp): {even_numbers}")
```
