## 3.5.1 Tensor hằng và Variable

TensorFlow có tensor hằng và tensor có thể thay đổi trạng thái (`tf.Variable`).

Trong Deep Learning, trọng số của mô hình thường được lưu trong `Variable`.

```python
x = tf.ones(shape=(2, 2))
y = tf.zeros(shape=(2, 2))
z = tf.random.normal(shape=(2, 2), mean=0.0, stddev=1.0)

print('Tensor x =')
print(x)
print('\nTensor y =')
print(y)
print('\nTensor z =')
print(z)
# Tạo một tensor 2x2 với các giá trị ngẫu nhiên phân bố đều 
# trong khoảng mặc định [0, 1).
v = tf.Variable(initial_value=tf.random.uniform(shape=(2, 2)))
print('\nVariable ban đầu =')
print(v)

v.assign(tf.ones((2, 2)))
print('\nVariable sau assign =')
print(v)

v.assign_add(tf.ones((2, 2)))
print('\nVariable sau assign_add =')
print(v)
```

## 3.5.2 Tensor operations

TensorFlow hỗ trợ nhiều phép toán quen thuộc như cộng, nhân ma trận, bình phương, căn bậc hai, giá trị lớn nhất theo từng phần tử.

Đây là nền tảng để xây dựng các phép biến đổi trong mạng nơ-ron.

```python
# Tạo hai tensor hằng (constant) dạng ma trận 2x2
a = tf.constant([[1.0, 2.0], [3.0, 4.0]])
b = tf.constant([[5.0, 6.0], [7.0, 8.0]])

# Phép cộng hai tensor (cộng từng phần tử)
print('a + b =')
print(a + b)

# Bình phương từng phần tử của tensor a
print('\ntf.square(a) =')
print(tf.square(a))

# Nhân ma trận (matrix multiplication) giữa a và b
print('\ntf.matmul(a, b) =')
print(tf.matmul(a, b))

# Lấy giá trị lớn hơn giữa a và b theo từng phần tử
print('\ntf.maximum(a, b) =')
print(tf.maximum(a, b))
```

## 3.5.3 GradientTape và tự động vi phân

`GradientTape` cho phép TensorFlow theo dõi các phép tính để tự động tính gradient.

Đây là cơ chế quan trọng trong huấn luyện mạng nơ-ron bằng gradient descent.

```python
# Khởi tạo biến w với giá trị ban đầu = 3.0 (có thể thay đổi trong quá trình tính toán)
w = tf.Variable(3.0)

# GradientTape dùng để ghi lại các phép toán nhằm phục vụ tính đạo hàm (gradient)
with tf.GradientTape() as tape:
    # Định nghĩa hàm y theo w (hàm bậc hai)
    y = w ** 2 + 2 * w + 1

# Tính đạo hàm của y theo w (dy/dw)
grad = tape.gradient(y, w)

# In ra giá trị của hàm y
print('Giá trị y =', y.numpy())

# In ra gradient (đạo hàm) tại w = 3
print('Gradient dy/dw =', grad.numpy())
```

## Ví dụ trực quan về Gradient Descent

Ta tối ưu một hàm đơn giản: `loss = (w - 4)^2`.

Mục tiêu là cập nhật `w` để giá trị tiến dần về 4.

```python
# Khởi tạo biến w với giá trị ban đầu = 0.0 (điểm xuất phát)
w = tf.Variable(0.0)

# Đặt learning rate (tốc độ học)
learning_rate = 0.1

# Lặp 10 bước để thực hiện gradient descent
for step in range(10):
    
    # Ghi lại các phép toán để tính gradient
    with tf.GradientTape() as tape:
        # Hàm loss cần tối ưu (mục tiêu là đưa w tiến về 4)
        loss = (w - 4.0) ** 2
    
    # Tính gradient của loss theo w (đạo hàm)
    grad = tape.gradient(loss, w)
    
    # Cập nhật w theo công thức gradient descent
    # w = w - learning_rate * gradient
    w.assign_sub(learning_rate * grad)
    
    # In thông tin mỗi bước: giá trị w, loss và gradient
    print(f'Bước {step:02d}: w = {w.numpy():.4f}, loss = {loss.numpy():.4f}, grad = {grad.numpy():.4f}')
```

## 3.6 Giải phẫu một mạng nơ-ron trong Keras

Các thành phần cốt lõi:

- `Layer`: khối xây dựng cơ bản.
- `Model`: tập hợp các layer.
- `compile()`: cấu hình optimizer, loss, metrics.
- `fit()`: huấn luyện mô hình.
- `evaluate()`: đánh giá.
- `predict()`: suy luận.

```python
# Import Keras từ TensorFlow
from tensorflow import keras

# Import module layers để xây dựng các lớp (layer) trong mạng neural
from tensorflow.keras import layers

# Khởi tạo mô hình Sequential (các layer xếp chồng tuần tự)
model = keras.Sequential([
    
    # Layer 1: Dense (fully connected)
    # 16 neuron, hàm kích hoạt ReLU
    # input_shape=(8,) nghĩa là mỗi mẫu dữ liệu có 8 đặc trưng đầu vào
    layers.Dense(16, activation='relu', input_shape=(8,)),
    
    # Layer 2 (output layer):
    # 1 neuron, dùng sigmoid cho bài toán phân loại nhị phân
    layers.Dense(1, activation='sigmoid')
])

# Hiển thị cấu trúc mô hình (số layer, số tham số, output shape...)
model.summary()
```

## Tạo dữ liệu giả để minh họa huấn luyện

Ở đây ta không dùng dữ liệu thật mà tạo dữ liệu nhị phân đơn giản để minh họa quy trình `compile()`, `fit()`, `evaluate()` và `predict()`.

```python
# Import thư viện NumPy để xử lý dữ liệu
import numpy as np

# Đặt seed để kết quả random có thể tái lập (reproducible)
np.random.seed(42)

# Tạo dữ liệu train gồm 200 mẫu, mỗi mẫu có 8 đặc trưng
x_train = np.random.random((200, 8)).astype('float32')

# Tạo nhãn (label) cho train:
# Nếu tổng các đặc trưng > 4 thì label = 1, ngược lại = 0
y_train = (x_train.sum(axis=1) > 4).astype('float32')

# Tạo dữ liệu test gồm 50 mẫu, mỗi mẫu có 8 đặc trưng
x_test = np.random.random((50, 8)).astype('float32')

# Tạo nhãn cho test theo cùng logic như train
y_test = (x_test.sum(axis=1) > 4).astype('float32')
```

```python
# Compile model: cấu hình cách học của mô hình
model.compile(
    
    # Optimizer: thuật toán tối ưu (cập nhật trọng số)
    # rmsprop là optimizer phổ biến cho Deep Learning
    optimizer='rmsprop',
    
    # Loss function: hàm mất mát
    # binary_crossentropy dùng cho bài toán phân loại nhị phân
    loss='binary_crossentropy',
    
    # Metrics: các chỉ số để đánh giá model trong quá trình train
    # accuracy = độ chính xác
    metrics=['accuracy']
)

# Huấn luyện mô hình
history = model.fit(
    
    # Dữ liệu đầu vào (features)
    x_train,
    
    # Nhãn tương ứng (labels)
    y_train,
    
    # Số lần lặp qua toàn bộ dữ liệu (epoch)
    epochs=5,
    
    # Số mẫu trong mỗi batch
    batch_size=16,
    
    # Tách 20% dữ liệu train để làm validation
    validation_split=0.2,
    
    # verbose=1: hiển thị tiến trình train
    verbose=1
)
```

```python
# Đánh giá mô hình trên tập test
# Trả về giá trị loss và accuracy
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)

# In ra loss (làm tròn 4 chữ số thập phân)
print('Test loss =', round(float(test_loss), 4))

# In ra độ chính xác trên tập test
print('Test accuracy =', round(float(test_acc), 4))

# Dự đoán xác suất cho 5 mẫu đầu tiên của tập test
pred = model.predict(x_test[:5])

# In ra kết quả dự đoán
print('\nDự đoán xác suất cho 5 mẫu đầu:')
print(pred)
```

## Quy tắc phân loại

Vì model dùng **sigmoid**, nên output là **xác suất nằm trong khoảng [0, 1]**.

**Quy tắc phổ biến**
Nếu xác suất ≥ 0.5 → dự đoán lớp 1  
Nếu xác suất < 0.5 → dự đoán lớp 0  

### Ví dụ

0.8 → lớp 1  
0.3 → lớp 0  
0.5 → thường quy về lớp 1  

### Giải thích bản chất

Sigmoid biến output thành xác suất của lớp 1  
P(y = 1) = output  
P(y = 0) = 1 - output  

### Lưu ý quan trọng

Ngưỡng 0.5 không phải lúc nào cũng tối ưu

Trong thực tế có thể điều chỉnh threshold:  
0.7 → giảm false positive  
0.3 → giảm false negative  

“Nếu model đoán 0.49 và 0.51 thì có khác nhau nhiều không?”
