# Artificial Intelligence - Week 02

## 1. Trích chọn đặc trưng - Feature Selection

**Trích chọn đặc trưng (Feature Selection)** là một trong những phần quan trọng cần lưu ý khi học về Trí tuệ nhân tạo. Lợi ích của việc Trích chọn đặc trưng là 
- Làm tăng tốc độ tính toán, giảm thiểu thời gian khi huấn luyện
- Đơn giản hóa mô hình và giúp chúng dễ diễn giải hơn
- Nâng cao hiệu suất bằng cách giảm thiểu dữ liệu đầu vào overfitting

Những tính năng thường bị loại bỏ khi trích chọn dặc trưng mà không làm ảnh hưởng tới hiệu suất / độ chính xác của mô hình thường thuộc vào một trong 2 loại sau:

- Dư thừa (Redundant): Đây là dạng tính năng có tính tương thích cao so với các tính năng đầu vào khác nên được bỏ qua mà không ảnh hưởng tới hiệu suất của mô hình.

- Không liên quan (Irrelevent): Đây là dạng tính năng có tương quan thấp với tính năng đầu vào nên được bỏ qua nếu không sẽ làm nhiễu mô hình hơn là hiệu quả.
