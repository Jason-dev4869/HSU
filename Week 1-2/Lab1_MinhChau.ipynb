a. Misising data could be anything in {NA,n/a,"",-1,...}. Read the data and assign the missing data in the dataframe as NaN. Hint: df.replace(...,np.nan,inplace=True)

```python
df = pd.read_csv('Lab1_dataset.csv')
df.head(10)
```

```python
df.replace(["NA", "n/a", "", -1], np.nan, inplace=True)
print(df.head(10))
```

b. Show the summary of the dataframe

```python
df.describe()
```

c. Count the number and percentage of missing data in each column and the entire dataset

```python
nan_counts = df.isna().sum()
nan_percent = nan_counts / len(df) * 100

# Missing data for each columns
nan_table_columns = pd.DataFrame({
    'Missing counts': nan_counts,
    'Missing percent (%)': nan_percent.round(2)
})

print("Missing table for each column:")
nan_table_columns
```

```python
# Missing data Entire Dataframe
nan_count_df = nan_counts.sum()
df_size = df.size
nan_percent_df = (nan_count_df / df_size) * 100

nan_table_df = pd.DataFrame({
    'Missing counts': [nan_count_df],
    'Missing percent (%)': [nan_percent_df.round(2)]
})

print("Missing table for entire DataFrame:")
nan_table_df
```

d. Randomly drop 25% of the rows that have missing data. Hint: use any() and numpy.random.choice()

```python
import numpy as np

missing_rows = df[df.isna().any(axis=1)].index

n_drop = int(0.25 * len(missing_rows))

drop_indices = np.random.choice(missing_rows, size=n_drop, replace=False)

df.drop(drop_indices, axis=0, inplace=True)

print(f"Số hàng đã xóa: {n_drop}")
print(df.shape)
```

e. Fill in the missing data with *median* (if the column data type is numeric) or *mode* (if the column data type is not numeric)

```python
#df colums type
df.dtypes
```

```python
fill_values = {}
for col in df.columns:
    if np.issubdtype(df[col].dtypes, np.number):
        fill_values[col] = df[col].median()
    else:
        fill_values[col] = df[col].mode()[0]

df.fillna(fill_values, inplace=True)
df.head(10)
```
