### Exercise 1. Greedy Best-First Search vs A* Search

Answer the following:

1. What is the evaluation function used in Greedy Best-First Search?
2. What is the evaluation function used in A* Search?
3. What is the role of `g(n)` in A* Search?
4. Why can Greedy Best-First Search fail to find an optimal solution?
5. Why can A* Search still reconsider paths that Greedy Best-First Search may ignore?


#### 1. What is the evaluation function used in Greedy Best-First Search?

The evaluation function used in Greedy Best-First Search is typically a heuristic function, denoted as h(n), which estimates the cost to reach the goal from node n. The algorithm selects the node with the lowest h(n) value to expand next, aiming to find a path to the goal as quickly as possible.

#### 2. What is the evaluation function used in A* Search?

The evaluation function used in A* Search is typically denoted as f(n) and is defined as the sum of two components: g(n) and h(n).
- g(n) represents the cost from the start node to the current node n. It is the actual cost incurred to reach node n from the starting point.
- h(n) represents the heuristic estimate of the cost from node n to the goal node. It is an estimate of the remaining cost to reach the goal from node n.

Therefore, the evaluation function can be expressed as:
**f(n) = g(n) + h(n)**

In A* Search, the algorithm selects the node with the lowest f(n) value for expansion, which allows it to efficiently find the optimal path to the goal while considering both the cost incurred so far and the estimated cost to reach the goal.

#### 3. What is the role of `g(n)` in A* Search?

`g(n)` represents the cost from the start node to the current node `n`. It is used in A* Search to calculate the total estimated cost `f(n)`, which is the sum of `g(n)` and `h(n)`, where `h(n)` is the heuristic estimate of the cost from node `n` to the goal. 

The A* algorithm uses `f(n)` to prioritize which nodes to explore next, aiming to find the most efficient path to the goal.

#### 4. Why can Greedy Best-First Search fail to find an optimal solution?

Greedy Best-First Search can fail to find an optimal solution because it relies solely on a heuristic function to guide its search. The algorithm selects the node that appears to be closest to the goal based on the heuristic, without considering the actual cost of reaching that node from the start. 

This can lead to situations where the algorithm gets stuck in a local minimum or chooses a path that seems promising but ultimately leads to a suboptimal solution. Additionally, if the heuristic is not well-designed or is misleading, it can further exacerbate this issue, causing the algorithm to overlook better paths that may have a higher initial cost but ultimately lead to an optimal solution.

#### 5. Why can A* Search still reconsider paths that Greedy Best-First Search may ignore?

A* Search can reconsider paths that Greedy Best-First Search may ignore because A* Search uses a combination of the actual cost from the start node to the current node (g(n)) and the heuristic estimate of the cost from the current node to the goal (h(n)). 

This allows A* Search to evaluate paths based on both their actual cost and their estimated cost to reach the goal, which can lead to reconsidering paths that may have been ignored by Greedy Best-First Search, which only considers the heuristic estimate (h(n)) and may overlook paths that have a higher heuristic value but a lower actual cost.

#### 2. What value would Greedy Best-First Search use to rank this node?

Greedy Best-First Search uses the heuristic function `h(n)` to rank nodes. Therefore, it would use the value of `h(n)`, which is 5, to rank this node.

#### 3. If another node `m` has `g(m) = 3` and `h(m) = 8`, which node would A* prefer?

A* would prefer node `m` because it has a lower total cost `f(m) = g(m) + h(m) = 3 + 8 = 11` compared to node `n` which has `f(n) = g(n) + h(n) = 7 + 5 = 12`.

#### 4. Which node would Greedy Best-First Search prefer?

Greedy Best-First Search would prefer the node with the lowest heuristic value `h(n)`. In this case, it would prefer node `n` because `h(n) = 5`, which is lower than any other node's heuristic value.

The two algorithms, A* and Greedy Best-First Search, may make different choices because they use different evaluation functions to determine which node to expand next.

### Exercise 4. Weighted A* Search

Weighted A* uses the evaluation function:

`f(n) = g(n) + W * h(n)`

Answer the following:

1. What happens when `W = 1`?
2. What happens when `W = 0`?
3. What happens conceptually when `W` becomes very large?
4. Why can Weighted A* often expand fewer nodes?
5. Why does Weighted A* lose the optimality guarantee when `W > 1`?


#### 1. What happens when `W = 1`?

When `W = 1`, the evaluation function simplifies to:
`f(n) = g(n) + h(n)`

This is the standard A* search algorithm, which combines the actual cost from the start node to the current node (`g(n)`) with the heuristic estimate of the cost from the current node to the goal node (`h(n)`). In this case, the search will be optimal and complete, as it will always expand the node with the lowest total cost.

#### 2. What happens when `W = 0`?

When `W = 0`, the evaluation function simplifies to `f(n) = g(n)`. This means that the algorithm will only consider the cost from the start node to the current node (g(n)) and will ignore the heuristic estimate of the cost to reach the goal (h(n)). 

As a result, the algorithm will behave like Dijkstra's algorithm, which explores all possible paths based on their actual costs without any guidance from a heuristic. This can lead to inefficient search, especially in large search spaces, as it may explore many unnecessary nodes before reaching the goal.

#### 3. What happens conceptually when `W` becomes very large?

When `W` becomes very large, the evaluation function `f(n)` becomes dominated by the heuristic component `W * h(n)`. This means that the search algorithm will heavily prioritize nodes that are estimated to be closer to the goal, as indicated by the heuristic function `h(n)`. As a result, the search may become more greedy, potentially leading to faster solutions in some cases. 

However, it can also lead to suboptimal solutions or even failure to find a solution if the heuristic is not accurate, as the algorithm may overlook paths that have a lower actual cost `g(n)` but a higher heuristic estimate. In extreme cases, if `W` is excessively large, the algorithm may behave like a pure greedy search, which can be inefficient and may not guarantee finding the optimal solution.

#### 4. Why can Weighted A* often expand fewer nodes?

Weighted A* can often expand fewer nodes because it places more emphasis on the heuristic estimate (h(n)) by multiplying it with a weight (W). This means that the algorithm is more likely to prioritize nodes that are estimated to be closer to the goal, potentially leading to a faster path to the solution. By focusing on promising paths, Weighted A* can avoid exploring less relevant nodes, thus reducing the total number of nodes expanded.

#### 5. Why does Weighted A* lose the optimality guarantee when `W > 1`?

Weighted A* loses the optimality guarantee when `W > 1` because it places more emphasis on the heuristic estimate `h(n)` than on the actual cost `g(n)`. When `W` is greater than 1, the algorithm may prioritize nodes that appear to be closer to the goal based on the heuristic, even if they have a higher actual cost. 

This can lead to situations where the algorithm selects a suboptimal path that seems promising due to the inflated heuristic value, rather than exploring paths that may have a lower total cost. As a result, Weighted A* with `W > 1` may not always find the optimal solution.

#### 1. What does it mean to build a heuristic from a subproblem?

Building a heuristic from a subproblem involves creating an approximation or a simplified version of the original problem to guide the search for a solution. 

This is often done by identifying a smaller, more manageable portion of the problem that can be solved more easily, and then using the insights gained from that subproblem to inform the decision-making process in the larger problem. 

The heuristic can help prioritize certain paths or actions based on the results obtained from solving the subproblem, ultimately leading to a more efficient search for the optimal solution in the original problem.

#### 2. Why can the exact cost of solving a subproblem be used as an admissible heuristic for the full problem?

The exact cost of solving a subproblem can be used as an admissible heuristic for the full problem because it provides a lower bound on the cost to reach the goal from the current state. 

An admissible heuristic is one that never overestimates the true cost to reach the goal, and since the exact cost of solving a subproblem is a precise measure of the cost to solve that specific part of the problem, it will always be less than or equal to the actual cost of solving the entire problem. 

Therefore, using the exact cost of a subproblem as a heuristic ensures that it remains admissible and can effectively guide search algorithms towards optimal solutions.

#### 3. What is a pattern database?

A pattern database is a precomputed table that stores the optimal solutions or heuristic values for specific subproblems of a larger problem. 

It is often used in search algorithms to improve efficiency by providing quick access to previously computed results, allowing the algorithm to make informed decisions and avoid redundant calculations. 

Pattern databases are commonly used in combinatorial optimization problems, such as puzzles and games, where they can significantly reduce the search space and improve performance.

#### 4. Why are pattern databases useful in search problems such as the 8-puzzle?

Pattern databases are useful in search problems like the 8-puzzle because they provide precomputed heuristic values that can significantly speed up the search process. A pattern database is a lookup table that contains the optimal number of moves required to solve a specific subproblem of the larger problem. 

In the case of the 8-puzzle, a pattern database can store the minimum number of moves required to arrange a subset of the tiles in their correct positions.By using a pattern database, the search algorithm can quickly retrieve heuristic values for different states of the puzzle, allowing it to make informed decisions about which paths to explore. 

This can lead to a significant reduction in the number of states that need to be evaluated, ultimately improving the efficiency of the search and reducing the time it takes to find a solution.

#### 5. Give one advantage and one limitation of subproblem-based heuristics.

**Advantage**: Subproblem-based heuristics can break down a complex problem into smaller, more manageable parts, allowing for more efficient problem-solving and potentially faster solutions.

**Limitation**: However, subproblem-based heuristics may not always capture the global optimal solution,as they focus on solving individual subproblems without considering the overall problem context, which can lead to suboptimal results.

````python
#Helper function
# to reconstruct the path from start to goal
def reconstruct_path(came_from: dict[str, str], 
                    current: str) -> list[str]:
    total_path = [current]
    while current in came_from:
        current = came_from[current]
        total_path.append(current)
    return total_path[::-1]  # reverse the path

# to compute the total cost of a path
def compute_cost(path: list[str]
, graph: dict[str, dict[str, int]]) -> int:
    return sum(graph[path[i]][path[i + 1]] 
        for i in range(len(path) - 1))

#to print results clearly
def print_report(name: str, path: list[str]
, cost: int, expanded: int) -> None:
    print(f"\n{'═'*50}")
    print(f"  {name}")
    print(f"{'═'*50}")
    print(f"  Path     : {' → '.join(path)}")
    print(f"  Cost     : {cost} km")
    print(f"  Expanded : {expanded} nodes")
````

#### Task 10.2 - Greedy Best-First Search

Implement Greedy Best-First Search using:

`f(n) = h(n)`

Your function should return at least:
- path
- total cost
- expanded nodes count

````python
#Greedy Best-First Search implementation
#using f(n) = h(n)
def greedy_best_first_search(graph, heuristic, start, goal):
    open_set = []
    heappush(open_set, (heuristic[start], start))
    came_from = {}
    expanded_nodes = set()

    while open_set:
        _, current = heappop(open_set)

        if current == goal:
            path = reconstruct_path(came_from, current)
            cost = compute_cost(path, graph)
            print_report("Greedy Best-First Search"
                , path, cost, len(expanded_nodes))
            return path

        expanded_nodes.add(current)

        for neighbor in graph[current]:
            if neighbor not in expanded_nodes:
                heappush(open_set
                    , (heuristic[neighbor], neighbor))
                if neighbor not in came_from:
                    came_from[neighbor] = current

    print_report("Greedy Best-First Search",
         [], 0, len(expanded_nodes))
    return None
````

#### Task 10.3 - A* Search

Implement A* Search using:

`f(n) = g(n) + h(n)`

Your function should return at least:
- path
- total cost
- expanded nodes count

````python
#implement A* Search
#using f(n) = g(n) + h(n)
def a_star_search(graph, heuristic, start, goal):
    open_set = []
    heappush(open_set, (heuristic[start], start))
    came_from = {}
    g_score = {node: float('inf') for node in graph}
    g_score[start] = 0
    expanded_nodes = set()

    while open_set:
        _, current = heappop(open_set)

        if current == goal:
            path = reconstruct_path(came_from, current)
            cost = compute_cost(path, graph)
            print_report("A* Search",
                 path, cost, len(expanded_nodes))
            return path

        expanded_nodes.add(current)

        for neighbor, cost in graph[current].items():
            tentative_g_score = g_score[current] + cost
            if tentative_g_score < g_score[neighbor]:
                came_from[neighbor] = current
                g_score[neighbor] = tentative_g_score
                f_score = tentative_g_score + heuristic[neighbor]
                heappush(open_set, (f_score, neighbor))

    print_report("A* Search", [], 0, len(expanded_nodes))
    return None
````
