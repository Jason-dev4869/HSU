## 1. Huấn luyện perceptron

Huấn luyện perceptron giải bài toán AND

```python
import numpy as np

# ==============================
# 1. Chuẩn bị dữ liệu bài toán AND
# ==============================

# Input: 2 biến nhị phân
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Nhãn AND tương ứng
y = np.array([0, 0, 0, 1])

# ==============================
# 2. Khởi tạo tham số perceptron
# ==============================

np.random.seed(42)

# Khởi tạo trọng số ngẫu nhiên (2 input → 2 trọng số)
w = np.random.randn(2)

# Bias
b = np.random.randn()

# Learning rate
lr = 0.1

# Số epoch
# Một epoch là một lần mô hình “nhìn thấy toàn bộ tập dữ liệu huấn luyện”
# VD: Có 100 mẫu dữ liệu:
#    - 1 epoch = mô hình học qua đủ 100 mẫu một lần
#    - 10 epoch = mô hình học qua 100 mẫu đó 10 lần
epochs = 10

# ==============================
# 3. Định nghĩa hàm kích hoạt (step)
# ==============================

def step_function(z):
    # Nếu z >= 0 trả về 1, ngược lại trả về 0
    return 1 if z >= 0 else 0

# ==============================
# 4. Huấn luyện perceptron
# ==============================

for epoch in range(epochs):
    print(f"\nEpoch {epoch+1}")
    
    for i in range(len(X)):
        xi = X[i]
        target = y[i]
        
        # Tính đầu ra tuyến tính
        z = np.dot(w, xi) + b
        
        # Qua hàm kích hoạt
        y_pred = step_function(z)
        
        # Sai số
        error = target - y_pred
        
        # Cập nhật trọng số và bias
        w = w + lr * error * xi
        b = b + lr * error
        
        print(f"Input: {xi}, Target: {target}, Pred: {y_pred}, Error: {error}")

# ==============================
# 5. Kiểm tra kết quả
# ==============================

print("\n=== Kết quả sau huấn luyện ===")

for i in range(len(X)):
    xi = X[i]
    z = np.dot(w, xi) + b
    y_pred = step_function(z)
    
    print(f"Input: {xi} → Output: {y_pred}")
```

## 2. Các hàm kích hoạt thông dụng

### 2.1. Hàm ReLU (Rectified Linear Unit)

```python
import numpy as np
import matplotlib.pyplot as plt

# Định nghĩa hàm ReLU
def relu(x):
    return np.maximum(0, x)  # lấy max giữa 0 và x

# Tạo dữ liệu đầu vào từ -5 đến 5
x = np.linspace(-5, 5, 100)

# Tính giá trị ReLU tương ứng
y = relu(x)

# Thêm tiêu đề và nhãn trục
plt.title("Đồ thị hàm ReLU")
plt.xlabel("x")
plt.ylabel("ReLU(x)")

# Vẽ trục hoành và trục tung
plt.axhline(0)
plt.axvline(0)

# Vẽ đồ thị
plt.plot(x, y, color='red')

# Hiển thị đồ thị
plt.show()
```

- Đoạn x < 0 → đồ thị nằm trên trục x (giá trị = 0)
- Đoạn x ≥ 0 → đồ thị là đường thẳng y = x
- Thường làm hàm kích hoạt cho các lớp ẩn

### 2.2. Hàm Tanh

```python
import numpy as np
import matplotlib.pyplot as plt

# Định nghĩa hàm tanh
def tanh(x):
    return np.tanh(x)  # sử dụng hàm có sẵn của numpy

# Tạo dữ liệu đầu vào từ -5 đến 5
x = np.linspace(-5, 5, 100)

# Tính giá trị tanh
y = tanh(x)

# Thêm tiêu đề và nhãn trục
plt.title("Đồ thị hàm tanh")
plt.xlabel("x")
plt.ylabel("tanh(x)")

# Vẽ trục tọa độ
plt.axhline(0)
plt.axvline(0)

# Vẽ đồ thị
plt.plot(x, y, color='red')

# Hiển thị đồ thị
plt.show()
```

**Công thức:**

$tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}}$

**Đặc điểm:**

- Giá trị nằm trong khoảng (-1, 1)
- Đối xứng qua gốc tọa độ
- Thường dùng ở hidden layer

### 2.3. Hàm Sigmoid

```python
import numpy as np
import matplotlib.pyplot as plt

# Định nghĩa hàm Sigmoid
def sigmoid(x):
    return 1 / (1 + np.exp(-x))  # công thức sigmoid

# Tạo dữ liệu đầu vào từ -10 đến 10
x = np.linspace(-10, 10, 100)

# Tính giá trị Sigmoid
y = sigmoid(x)

# Thêm tiêu đề và nhãn trục
plt.title("Đồ thị hàm Sigmoid")
plt.xlabel("x")
plt.ylabel("Sigmoid(x)")

# Vẽ trục tọa độ
plt.axhline(0)
plt.axvline(0)

# Vẽ đồ thị với màu đỏ
plt.plot(x, y, color='red')

# Hiển thị đồ thị
plt.show()
```

- Đồ thị có dạng chữ S
- Giá trị nằm trong khoảng (0, 1)
- Thường dùng ở tầng output cho bài toán phân loại nhị phân

### 2.4. Hàm Softmax

Vẽ đồ thị Softmax (3 lớp)

```python
import numpy as np
import matplotlib.pyplot as plt

# Hàm Softmax cho vector x
def softmax(x):
    # Trừ max(x) để tránh tràn số (numerical stability)
    exp_x = np.exp(x - np.max(x))
    return exp_x / np.sum(exp_x)

# Tạo giá trị biến thiên cho 1 phần tử
x_vals = np.linspace(-5, 5, 100)

# Giữ 2 phần tử cố định, thay đổi phần tử thứ 1
y1 = []
y2 = []
y3 = []

for x in x_vals:
    vec = np.array([x, 1.0, 0.5])  # vector đầu vào
    s = softmax(vec)
    y1.append(s[0])
    y2.append(s[1])
    y3.append(s[2])

# Vẽ đồ thị (3 đường tương ứng 3 lớp)
plt.plot(x_vals, y1, label="Class 1")
plt.plot(x_vals, y2, label="Class 2")
plt.plot(x_vals, y3, label="Class 3")

# Tiêu đề và nhãn
plt.title("Đồ thị hàm Softmax")
plt.xlabel("Giá trị đầu vào thay đổi")
plt.ylabel("Xác suất")

# Hiển thị chú thích
plt.legend()

# Hiển thị
plt.show()
```

**Công thức:**

$Softmax(x_i) = \frac{e^{x_i}}{\sum_{j} e^{x_j}}$

**Ý nghĩa:**
- Biến vector thành phân phối xác suất
- Tổng các giá trị = 1
- Giá trị lớn hơn → xác suất cao hơn
Ứng dụng
- Dùng ở tầng output của bài toán phân loại nhiều lớp (multi-class classification)

## 3. Tập dữ liệu MNIST

```python
# =========================
# 1. Import thư viện cần thiết
# =========================
from tensorflow.keras.datasets import mnist
import matplotlib.pyplot as plt

# =========================
# 2. Load dữ liệu MNIST
# =========================
# x_train: ảnh train, y_train: nhãn train
# x_test, y_test: dữ liệu test
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# =========================
# 3. In thông tin dữ liệu
# =========================
print("Số lượng ảnh train:", x_train.shape)
print("Số lượng ảnh test:", x_test.shape)

# =========================
# 4. Hiển thị 5 ảnh đầu tiên
# =========================
for i in range(5):
    plt.subplot(1, 5, i + 1)   # tạo lưới 1 hàng 5 cột
    plt.imshow(x_train[i], cmap='gray')  # hiển thị ảnh grayscale
    plt.title(f"Label: {y_train[i]}")    # hiển thị nhãn
    plt.axis('off')                     # tắt trục

plt.show()

# =========================
# 5. In giá trị pixel của ảnh đầu tiên
# =========================
print("Ma trận pixel của ảnh đầu tiên:")
print(x_train[0])
```

## 4. Nhận dạng chữ số viết tay MNIST

```python
# ==============================
# 1. Import thư viện cần thiết
# ==============================
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.datasets import mnist

# ==============================
# 2. Load dữ liệu MNIST
# ==============================
# train_images: dữ liệu huấn luyện (60000 ảnh)
# train_labels: nhãn tương ứng
# test_images: dữ liệu kiểm tra (10000 ảnh)
# test_labels: nhãn kiểm tra
(train_images, train_labels), (test_images, test_labels) = mnist.load_data()

# ==============================
# 3. Xây dựng mô hình mạng neural
# ==============================
# Sequential: mô hình xếp chồng các layer
model = keras.Sequential([
    # Layer đầu vào: 28x28 = 784 đặc trưng
    keras.Input(shape=(28*28,)),
    layers.Dense(512, activation="relu"),
    # Layer đầu ra: 10 lớp (0–9), dùng softmax
    layers.Dense(10, activation="softmax")
])

# ==============================
# 4. Compile mô hình
# ==============================
# optimizer: thuật toán tối ưu
# loss: hàm mất mát
# metrics: chỉ số đánh giá
model.compile(
    optimizer="rmsprop",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# ==============================
# 5. Tiền xử lý dữ liệu
# ==============================
# reshape ảnh từ (28,28) -> (784,)
train_images = train_images.reshape((60000, 28 * 28))
test_images = test_images.reshape((10000, 28 * 28))

# chuẩn hóa dữ liệu về [0,1]
train_images = train_images.astype("float32") / 255
test_images = test_images.astype("float32") / 255

# ==============================
# 6. Huấn luyện mô hình
# ==============================
model.fit(
    train_images,
    train_labels,
    epochs=5,        # số lần lặp qua toàn bộ dữ liệu
    batch_size=128   # số mẫu mỗi batch
)

# ==============================
# 7. Dự đoán trên dữ liệu mới
# ==============================
test_digits = test_images[0:10]   # lấy 10 ảnh đầu
predictions = model.predict(test_digits)

# in xác suất dự đoán của ảnh đầu tiên
print("Xác suất dự đoán:", predictions[0])

# lớp dự đoán (lớp có xác suất cao nhất)
predicted_label = predictions[0].argmax()
print("Dự đoán là số:", predicted_label)

# so sánh với nhãn thật
print("Nhãn thật:", test_labels[0])

# ==============================
# 8. Đánh giá mô hình
# ==============================
test_loss, test_acc = model.evaluate(test_images, test_labels)

print("Độ chính xác trên tập test:", test_acc)
```
