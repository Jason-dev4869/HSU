## 1. Khái niệm về Trí tuệ nhân tạo (Artificial Intelligence)

Ngày 31 tháng 8 năm 1955, nhà khoa học máy tính người Mỹ - John McCathy (1927 - 2011) lần đầu tiên đặt ra khái niệm **Artificial Intelligence** trong đề xuất hội thảo Dartmouth - "A Proposal for the Dartmouth Summer Research Project on Artificial Intelligence" do ông đứng ra tổ chức cùng với những nhà tiên phong trong lĩnh vực này là Marvin Minsky, Nathaniel Rochester và Claude Shannon.

<img src='img/Dartmouth Conference.png' width=500 alt-text = " A Proposal for the Dartmouth Summer Research Project on Artificial Intelligence">

Tới mùa hè năm 1956, Khái niệm được biết rộng rãi hơn với công chúng trong buổi hội thảo Dartmouth và khai sinh ra lĩnh vực nghiên cứu khoa học, kỹ thuật về Trí tuệ nhân tạo

<img src='img/JohnMcCathy_1956_Stanford.png' width= 500>

**Trí tuệ nhân tạo (Artificial Intelligence)** được hiểu là khả năng tư duy và học tập của một phần mềm / chương trình máy tính / máy móc do con người tạo ra nhằm thực hiện các nhiệm vụ phức tạp và có tính sáng tạo cao, tương tự như con người

*Q: Vậy làm thể nào để nhận biết rằng ở mức độ nào thì Máy móc được cho là có thể tư duy được ?*

*A: Phép thử Turing (Turing Test) là một trong những phương pháp được sử dụng để đánh giá khả năng tư duy của một chương trình máy tính.*

Phép thử Turing (Turing Test) được đưa ra bởi người tiên phong trong lĩnh vực khoa học máy tính - nhà toán học người Anh Alan Turing (1912 - 1954) vào năm 1950 trong bài viết "Máy tính và Trí tuệ" (Computing Machinery and Intelligence)
- Mục đích của phép thử là đưa ra một cách thực tiễn đánh giá khả năng tư duy của một chiếc máy tính để trả lời cho câu hỏi "Máy móc có thể tư duy được không ?" 
- Từ mục đích ấy mà trở thành một tiêu chuẩn đánh giá để hình thành nên lĩnh vực **Trí tuệ nhân tạo** sau này 

<img src="img/AlanTuring_n_EnigmaMachine .png" width="500" alt-text = "Alan Turing và Máy Enigma">

Turing test / Imitation Game:  Một người (A) thực hiện một cuộc thảo luận bằng **ngôn ngữ tự nhiên** với một con người khác (B) và một cái máy tính (Cm). Ba bên được cách ly với nhau, B và Cm phải cố gắng chứng tỏ mình là một con người.
- Nếu A không thể phân biệt được người và máy, thì Cm được coi là có khả năng tư duy như một con người. *[Vượt qua được phép thử]*
- Nếu A có thể phân biệt được người và máy, thì Cm được coi là không có khả năng tư duy như một con người. *[Không vượt qua được phép thử]*

## 2. Quy trình hoạt động và các nhánh của Trí tuệ nhân tạo

Sơ đồ tập con của AI:

Các nhánh của Trí tuệ nhân tạo gồm có:
- Machine Learning & Pattern Recognition
    + Định nghĩa: Học từ dữ liệu để đưa ra dự đoán quyết định và khả năng nhận diện mẫu trong dữ liệu
    + Kỹ thuật: 
        + Supervised Learning
        + Unsupervised Learning
        + Semi-supervised Learning
        + Reinforcement Learning
        + Neural Networks
    + Ứng dụng: Nhận diện khuôn mặt, phân loại ảnh, phát hiện gian lận, dịch máy, v.v..

- Logic-based AI:
    + Định nghĩa: Dựa vào logic hình thức để mô hình hóa suy luận, đưa ra quyết định dựa trên các quy tắc chung
    + Kỹ thuật:
        + Prolog
        + Rule-based Systems
        + First-Order Logic
        + Expert Systems
    + Ứng dụng: Hệ thống hỗ trợ quyết định, hệ thống quản lý dữ liệu, hệ thống cảnh báo, chatbot dựa trên quy tắc, suy luận pháp lý , v.v..

- Search
    + Định nghĩa: Tìm kiếm lời giải, giải pháp trong không gian, trạng thái để đi tìm tới lời giải, kết quả
    + Kỹ thuật:
        + Best-First Search (BFS)
        + Depth-First Search (DFS)
        + A* Search
        + Beam Search
        + Minimax
    + Ứng dụng: Tìm kiếm thông tin, tìm kiếm đường đi, tìm kiếm giải pháp tối ưu, v.v

- Knowledge Representation
    + Định nghĩa: Trình bày tri thức cho phép máy lưu trữ, truy vấn, suy luận 
    + Kỹ thuật:
        + Ontologies
        + Semantic Networks
        + Frames
        + Knowledge Graphs
    + Ứng dụng: Trợ lý ảo, chatbox thông minh, hệ thống trả lời tự động, v.v.

- Planning
    + Định nghĩa: Tạo ra chuỗi hành động hợp lý để đạt mục tiêu trong môi trường có điều kiện
    + Kỹ thuật:
        + STRIPS
        + Hierarchical planning
        + Partial-order planning
        + Planning as Satisfiability (SAT)
    + Ứng dụng: Robot di chuyển, lập kế hoạch tài chính, sắp xếp lịch trình, game chiến thuật, v.v.

- Heuristics
    + Định nghĩa: Quy tắc gần đúng giúp tìm ra giải pháp nhanh, chấp nhận kết quả không tối ưu, chưa đủ tốt (Phỏng đoán)
    + Kỹ thuật:
        + A* Heuristic Function
        + Hill Climbing
        + Evalution Function
    + Ứng dụng: Tăng hiệu quả thuật toán tìm kiếm, ra quyết định khi không đủ thời gian

- Genetic Programming
    + Định nghĩa: Phương pháp tính toán dựa trên cơ chế tiến hóa dùng quy tắc chọn lọc tự nhiên để tìm ra giải pháp tối ưu
    + Kỹ thuật:
        + Mutation Crossover
        + Fitness Function
        + Selection
    + Ứng dụng: Tối ưu hóa, tự động hoá lập trình, thiết kế thuật toán, thiết kế mạch, v.v.

Quy trình hoạt động của mô hình máy học (Machine Learning):

<img src='img/MLpipelines.png' width=500>

1. **Xác định vấn đề**: Xác định vấn đề mà bạn muốn giải quyết bằng mô hình
2. **Thu thập dữ liệu**: Thu thập dữ liệu liên quan đến vấn đề
3. **Chuẩn bị dữ liệu**: Chuẩn bị dữ liệu cho mô hình, bao gồm việc xử lý dữ liệu, loại bỏ dữ liệu không cần thiết, và chuyển đổi dữ liệu thành dạng phù hợp
4. **Chọn mô hình**: Chọn mô hình phù hợp với vấn đề và dữ liệu
5. **Đào tạo mô hình**: Đào tạo mô hình bằng cách sử dụng dữ liệu đã chuẩn bị
6. **Đánh giá mô hình**: Đánh giá hiệu suất của mô hình bằng cách sử dụng dữ liệu thử nghiệm
7. **Cải thiện mô hình**: Cải thiện hiệu suất của mô hình bằng cách sử dụng các kỹ thuật như tăng cường dữ liệu, giảm thiểu dữ liệu, và thử nghiệm các mô hình khác nhau
