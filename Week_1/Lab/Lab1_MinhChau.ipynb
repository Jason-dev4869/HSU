```python
import pandas as pd
import string

df = pd.read_csv('dataset.csv', header=None)
df.columns = list(string.ascii_uppercase[:df.shape[1]])

print(df.head())
```

a. Find the number of missing data in all columns. What’s the missing percentage? Hint: use df.isnull() or df.isna(). Show the first 10 lines of the output

```python
missing_count = df.isnull().sum()
missing_percent = (missing_count / len(df)) * 100

miss_df = pd.DataFrame({
    'Missing Count': missing_count,
    'Missing Percent (%)': missing_percent
})

print(miss_df)
```

b. Use the original data. Fill in the missings with [column] mean.  Hint: fillna(). Show the first 10 lines of the output.

c. Use the original data. Fill in the missings with [column] median. Show the first 10 lines of the output. 

```python
df_filled_median = df.fillna(df.median())
print(df_filled_median.head(10))
```

d. Use the original data. Fill in the missings with [column] mode. Show the first 10 lines of the output. 

```python
for col in df.columns:
    mode_series = df[col].mode()
    if not mode_series.empty:
        mode_value = mode_series[0]
        df[col] = df[col].fillna(mode_value)

print(df.head(10))
```

e. What is the difference between mean, median and mode?

- Mean: The mean is the average of all the numbers in a dataset. It is calculated by adding up all the numbers and then dividing by the total count of numbers.

- Median: The median is the middle value in a dataset when it is ordered from smallest to largest. If there are an even number of observations, the median is the average of the two middle numbers.

- Mode: The mode is the value that appears most frequently in a dataset. A dataset can have more than on mode if there are multiple values that appear with the same frequency and are more frequent than any other value.

f. Use the original data. Drop columns with a missing value rate higher than 30%. Show the first 10 lines of the output. 

```python
miss_ratio_col = ((df.isna().mean()) * 100).round(2)
print("Missing Ratio by Column:")
print(miss_ratio_col.to_frame(name="Missing %"))
```

```python
keep_col = []
for col in df.columns:
    if miss_ratio_col[col] >= 30.0:
        keep_col.append(col)
        
df_drop = df[keep_col]
print(df_drop.head(10))
```

g. Use the original data. Drop rows with missing value rate higher than 40%. Show the first 
10 lines of the output. 

```python
miss_ratio_row = ((df.isna().mean(axis=1)) * 100).round(2)
print("Missing Ratio by Row:")
print(miss_ratio_row.to_frame(name="Missing %"))
```

```python
df_drop_row = df[miss_ratio_row >= 40]
print(df_drop_row.head(10))
```

h. Use the original data. Normalize data columns using minmax normaliza on. Show the first 10 lines of the output. 

```python
df_nom = df.copy()

for col in df_nom.columns:
    col_min = df_nom[col].min()
    col_max = df_nom[col].max()
    if pd.notnull(col_max) and col_max != col_min:
        df_nom[col] = (df_nom[col] - col_min) / (col_max - col_min)

print('Data after normalization:')
print(df_nom.head(10))
```

```python
col_A = df['A']
mean_A = col_A.mean()
std_A = col_A.std()

lower_bound = mean_A - 2 * std_A
upper_bound = mean_A + 2 * std_A

df_filtered = df[(col_A >= lower_bound) & (col_A <= upper_bound)]

print(f"Mean: {mean_A:.2f}")
print(f"Std: {std_A:.2f}")
print(f"Lower Bound: {lower_bound:.2f}")
print(f"Upper Bound: {upper_bound:.2f}")

print(f"After drop outlier row:")
print(df_filtered)
```
