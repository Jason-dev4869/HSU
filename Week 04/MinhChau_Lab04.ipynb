## Nguyễn Phúc Minh Châu - Lab04

Sử dụng dữ liệu anh thanh đính kèm (file đính kèm) thực hiện các yêu cầu sau.

Xây dựng mô 

- input: text các chữ số ("123", "422", ...)

- Chuyển đổi các chữ số thành phổ đồ âm thanh tương ứng: dựa vào phổ đồ của các âm thanh của từng chữ số (dữ liệu âm thanh đính kèm) sau đó ghép các phổ đồ âm thanh của từng ký tự thành 1 chuỗi.

- Chuyển đổi phổ đồ âm thành thành sóng âm thanh và phát âm thanh lên.
