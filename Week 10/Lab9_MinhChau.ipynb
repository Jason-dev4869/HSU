## Question 1: When calculating the **sample standard devation** (**or sample devation**), why do we divide by (N-1) instead of N ?

To get an **unbiased** estimate of the population standard deviation, we are trying to estimate **Population** (huge) variance based on a **Sample** (small) variance. So to account for that discrepancy, **N-1** is used instead of **N**
