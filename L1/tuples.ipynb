# Tuples

```python
my_tuple = ("apple", "banana", "cherry", "date")
print(my_tuple)
```

```python
print(my_tuple[1:3])  
```

```python
print(my_tuple[-1])   
```

```python
def adder_subtractor(num1, num2):
    return (num1 + num2, num1 - num2) # Returns a tuple

sum_val, diff_val = adder_subtractor(10, 5)
print(f"Sum: {sum_val}, Difference: {diff_val}") # Output: Sum: 15, Difference: 5
```
