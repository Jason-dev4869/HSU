### Exercise 2: Describe the tic-tac-toe problem using PEAS.
----

To describe the Tic-Tac-Toe problem using the PEAS (Performance, Environment, Actuators, Sensors) framework, we must look at the game from the perspective of an intelligent agent designed to play it.PEAS Breakdown for Tic-Tac-ToeP: 

- **Performance Measure (P)**: The metrics used to evaluate how well the agent is doing. In Tic-Tac-Toe, this is typically a win/loss/draw ratio. A high-performing agent maximizes wins and minimizes losses (or at least forces a draw in every game, as the game is mathematically "solvable").

- **Environment (E)**: The external world where the agent operates. For this problem, it is a $3 \times 3$ grid. The environment is discrete (limited number of states), static (the board doesn't change while the agent is "thinking"), and observable (the agent can see the entire board).

- **Actuators (A)**: The tools the agent uses to change the state of the environment. In a digital version, this would be the function that places an "X" or an "O" into an empty cell on the grid.

- **Sensors (S)**: The tools the agent uses to perceive the environment. This is the mechanism that "reads" the current board state to identify which cells are empty and which contain symbols.

### Exercise 3: Define the search problem of tic-tac-toe
----
1. **Initial State**: 
An empty $3 \times 3$ board. Usually, the state also tracks whose turn it is (e.g., "Player X to move").

2. **Actions**: 
A set of all legal moves available to the current player. Specifically: {Place [Symbol] in empty cell $(r, c)$}, where $r$ and $c$ are the row and column indices ranging from 1 to 3.

3. **Transition Model**:
A function that returns the board configuration resulting from the chosen action. For example, if the action is "Place X in $(1, 1)$," the transition model returns a new state where $(1, 1)$ is occupied by X and all other cells remain the same.

4. **Goal State**:
condition where the game ends. This occurs when:
    - One player has three of their symbols in a row (horizontally, vertically, or diagonally).
    - The board is completely full (a draw).

5. **Path Cost**:
In Tic-Tac-Toe, each move is typically assigned a constant step cost (e.g., 1). However, in adversarial search (games), we often focus on the "Utility" of the final state rather than the path cost—assigning values like $+1$ for a win, $0$ for a draw, and $-1$ for a loss.

### Exercise 4: Describe the Tower of Hanoi problem using PEAS
----
The Tower of Hanoi is a classic mathematical puzzle often used in AI and computer science to demonstrate recursion and state-space search. To describe it using the PEAS (Performance, Environment, Actuators, Sensors) framework, we treat the problem-solver as an autonomous agent tasked with moving the disks.

- **Performance Measure**: The agent aims to minimize the number of moves (the optimal solution for n disks is 2n−1). It must also ensure 100% adherence to legal move rules (no larger disk on a smaller one).

- **Environment**: "A set of three vertical pegs (Source, Auxiliary, Target) and a stack of n disks of varying sizes. The environment is static, discrete, and fully observable."

- **Actuators**: A robotic arm or a software function capable of picking up a disk from one peg and dropping a disk onto another peg.

- **Sensors**: A vision system or a digital state tracker that identifies the current position of every disk and the top-most disk available on each peg.

### Exercise 5: Define the search problem of Tower of Hanoi
----
1. **Initial State**: 
The starting configuration where all $n$ disks are stacked on the Source Peg (usually the leftmost peg) in increasing order of size from top to bottom.

2. **Actions**: 
The set of legal moves available to the agent. An action is defined as moving the top disk from one peg $P_i$ to another peg $P_j$, provided:
    - Peg $P_i$ is not empty.
    - The disk being moved is smaller than the top disk currently on $P_j$ (or $P_j$ is empty).

3. **Transition Model**: 
A function that returns the resulting state after an action is applied.
    - Result(State, Action) $\rightarrow$ New State
    - If the agent moves Disk 1 from Peg A to Peg C, the transition model updates the coordinates/indices of Disk 1 to reflect its new position while keeping all other disks constant.

4. **Goal State**: 
The target configuration where all $n$ disks have been successfully moved to the Target Peg (usually the rightmost peg), maintained in their original size order.

5. **Path Cost**: 
The total "weight" of the solution. In the Tower of Hanoi:
    - Each individual move has a cost of 1.
    - The total path cost is the sum of all moves.
    - The objective is to find the path where the total cost is minimized ($2^n - 1$).

State Space Complexity: Because every disk can be on any of the three pegs, there are $3^n$ possible valid states. As $n$ increases, the state space grows exponentially, which is why efficient search strategies (or recursive algorithms) are essential.

Describing the **Transition Function in tabular form for the 4-Puzzle problem**.

1. Brief description of the 4-Puzzle problem
   The 4-Puzzle consists of three numbered tiles and one blank tile arranged on a 2×2 grid. A state represents a particular arrangement of the tiles, where 0 denotes the blank. An action moves the blank tile one position up, down, left, or right, if the move is legal. The transition function defines the resulting state after applying an action to a given state.

2. State representation
   Each state is represented as a 2×2 matrix, read row by row from left to right and from top to bottom.

Example state S₁: [ 1 0 2 3 ], 0 represents the blank tile.

3. Action set:
   - Up: move the blank tile up
   - Down: move the blank tile down
   - Left: move the blank tile left
   - Right: move the blank tile right

If an action is not applicable, the transition is undefined.

4. Transition Function table (example for a specific state)

**Your task:** Fill the Transition Function table for the inital state S1 = [1 3 0 2]

To solve this, we first visualize the grid for the initial state $S_1 = [1, 3, 0, 2]$. Following the "row by row, left to right" rule, the $2 \times 2$ matrix is:

$$\begin{bmatrix} 1 & 3 \\ 0 & 2 \end{bmatrix}$$

- The blank tile ($0$) is at the bottom-left position $(1, 0)$.
- Up: The blank moves to $(0, 0)$, swapping with the tile $1$.
- Down: The blank cannot move further down (off the grid).
- Left: The blank cannot move further left (off the grid).Right: The blank moves to $(1, 1)$, swapping with the tile $2$.

Transition Function Table for $S_1 = [1, 3, 0, 2]$

| Current State | Action | Next State     |
| ------------- | ------ | -------------- |
| [1 3; 0 2]    | Up     | [0 3; 1 2]     |
| [1 3; 0 2]    | Down   | Not applicable |
| [1 3; 0 2]    | Left   | Not applicable |
| [1 3; 0 2]    | Right  | [1 3; 2 0]     |

Logic Explanation
1. **Up Action**: Moving the blank tile up swaps it with the number $1$. The new sequence becomes $0$ (top-left), $3$ (top-right), $1$ (bottom-left), and $2$ (bottom-right).

2. **Right Action**: Moving the blank tile to the right swaps it with the number $2$. The new sequence becomes $1$ (top-left), $3$ (top-right), $2$ (bottom-left), and $0$ (bottom-right).

3. **Invalid Moves**: Because the blank is in the bottom-left corner, any attempt to move Down or Left would result in going outside the boundaries of the $2 \times 2$ grid, making those transitions undefined.
